```python
# Paste this into Google Colab and run
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import parallel_coordinates

# ---- Constructed (evidence-backed draft) dataset ----
data = [
    ["Indonesia", 2023, 1, 1, 1, 1, 1],
    ["Malaysia", 2023, 1, 1, 1, 1, 2],
    ["Thailand", 2023, 1, 1, 1, 1, 2],
    ["Singapore", 2024, 2, 2, 2, 2, 2],
    ["Philippines", 2023, 1, 0, 1, 1, 1],
]
cols = ["country","year","standards","economic_instruments","infrastructure","MRV_transparency","recognition"]
df = pd.DataFrame(data, columns=cols)

# GPRSI calculations
df["GPRSI_raw_mean"] = df[["standards","economic_instruments","infrastructure","MRV_transparency","recognition"]].mean(axis=1)
df["GPRSI_0_100"] = df["GPRSI_raw_mean"] * 50
df["sum_score"] = df[["standards","economic_instruments","infrastructure","MRV_transparency","recognition"]].sum(axis=1)
df["GPRSI_minmax"] = (df["sum_score"] - df["sum_score"].min()) / (df["sum_score"].max() - df["sum_score"].min()) * 100

# Show table
print(df.to_markdown(index=False))

# Bar plot
plt.figure(figsize=(8,4))
plt.bar(df["country"], df["GPRSI_0_100"])
plt.title("GPRSI (0-100) by Country — constructed dataset")
plt.ylabel("GPRSI (0-100)")
plt.xlabel("Country")
plt.tight_layout()
plt.show()

# Parallel coordinates (normalized components 0-1)
df_plot = df[["country","standards","economic_instruments","infrastructure","MRV_transparency","recognition"]].copy()
for c in ["standards","economic_instruments","infrastructure","MRV_transparency","recognition"]:
    df_plot[c] = df_plot[c] / 2.0
plt.figure(figsize=(10,5))
parallel_coordinates(df_plot, "country", linewidth=1.5)
plt.title("Normalized component comparison (0-1 scale)")
plt.ylabel("Normalized score (0-1)")
plt.legend(loc='best')
plt.tight_layout()
plt.show()
```

```python
# Colab-ready notebook: Scrape policy pages (2018-2024) and build GPRSI
# Instructions: Run this notebook in Google Colab (internet-enabled). It will:
# 1) Download a curated set of policy pages / PDFs for Indonesia, Malaysia, Thailand, Singapore, Philippines, and regional/global items
# 2) Extract text and detect keyword evidence for policy pillars
# 3) Convert evidence into ordinal scores (0/1/2) per pillar per country-year (2018-2024)
# 4) Compute GPRSI and produce charts + CSV export

# NOTE: This is an automated, heuristic approach. You should review and adjust codings manually
# after running (especially for nuanced legal texts). The heuristic maps keyword hits -> ordinal scores.

# ----------------------------
# Install dependencies (run once)
# ----------------------------
!pip install --quiet requests beautifulsoup4 pdfplumber pandas matplotlib scikit-learn tqdm

# ----------------------------
# Imports
# ----------------------------
import requests
from bs4 import BeautifulSoup
import pdfplumber
import io
import re
import pandas as pd
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

# ----------------------------
# Curated source URLs (evidence seeds)
# Add more URLs as you find them.
# ----------------------------
sources = {
    'Singapore': [
        'https://www.mpa.gov.sg/maritime-singapore/sustainability/maritime-singapore-green-initiative',
        'https://www.mpa.gov.sg/media-centre/details/port-marine-circular-no.-13-of-2024-revisions-to-the-maritime-singapore-green-initiative-%28msgi%29---integration-of-green-port-programme-%28gpp%29-into-the-enhanced-green-ship-programme-%28gsp%29-for-ocean-going-vessels',
        'https://www.mpa.gov.sg/media-centre/details/port-marine-circular-no.-12-of-2024-implementation-of-singapore-standard-ss-648--2024---code-of-practice-for-bunker-mass-flow-metering',
        'https://www.reuters.com/world/asia-pacific/singapore-mandate-digital-bunker-notes-starting-april-2025-2024-10-09/',
        "https://www.mpa.gov.sg/web/portal/home/mpa-maritime-singapore/green-initiatives/green-port-programme",
        "https://www.mpa.gov.sg/docs/mpaorg/pdf/MPA_Annual_Report_2023.pdf",
        "https://sustainabledevelopment.un.org/content/documents/commitments/commitment_1537.pdf"
        "https://www.mpa.gov.sg/",   # Maritime and Port Authority
        "https://www.nea.gov.sg/",   # National Environment Agency
        "https://www.mti.gov.sg/",   # Ministry of Trade and Industry
    ],
    'Malaysia': [
        'https://www.pka.gov.my/phocadownload/policy/PORT%20KLANG%20GREEN%20POLICY.pdf',
        'https://www.transportevents.com/presentations/Manila2022/Port_Klang_Authority.pdf',
         "https://www.mot.gov.my/en/maritime/green-port",
        "https://www.pka.gov.my/index.php/en/green-initiatives",  # Port Klang green program
        "https://www.marine.gov.my/jlm/resources/user_1/Misc/pdf/MSM/Malaysia_Shipping_Masterplan_2017-2022.pdf"
        "https://www.doe.gov.my/",   # Department of Environment
        "https://www.mot.gov.my/",   # Ministry of Transport (green shipping policy)
        "https://www.mea.gov.my/",   # Ministry of Economic Affairs
    ],
    'Philippines': [
        'https://www.ppa.com.ph/content/Supporting%20Documents/Item%201.c_PPA%20AO%20No.%2005-2018%20%20The%20Port%20Environmental%20Policy%20%28PEP%29_0.pdf',
        'https://www.scribd.com/document/379966171/Philippine-Ports-Authority-AO-No-05-2018-Port-Environmental-Policy'
        "https://marina.gov.ph/wp-content/uploads/2022/01/MARINA-Green-Port-Initiatives.pdf",
        "https://www.marina.gov.ph/reports/",
        "https://denr.gov.ph/images/DENR_Digital_Annual_Report_2022.pdf"  # DENR annual report
        "https://www.denr.gov.ph/",  # Department of Environment and Natural Resources
        "https://www.ppa.com.ph/",   # Philippine Ports Authority
        "https://www.dti.gov.ph/",   # Department of Trade and Industry
    ],
    'Thailand': [
        'https://mddb.apec.org/Documents/2024/TPTWG/MEG-TM1/24_tptwg_meg_tm1_007.pdf',
        'https://pat.marketing/en/news/green-port/',
        "https://www.port.co.th/eng/sustainability",
        "https://www.md.go.th/statics/service/en/1147",  # Marine Department – Green Port Policy
        "https://www.oie.go.th/sites/default/files/attachments/industrial_environment/Green%20Industry%20Handbook%202020.pdf"
        "https://www.port.co.th/",   # Port Authority of Thailand
        "https://www.onep.go.th/",   # Office of Natural Resources & Environmental Policy
        "https://www.mot.go.th/",    # Ministry of Transport
    ],
    'Indonesia': [
        'https://peraturan.bpk.go.id/Details/103722/permenhub-no-29-tahun-2014',
        'https://budidjaja.law/2023/09/green-shipping-how-should-indonesia-react/',
        "https://www.dephub.go.id/post/read/peraturan-menteri-perhubungan-no-29-tahun-2018",   # Green Port Guidelines
        "https://jdih.dephub.go.id/produk_hukum/view/VFVGNQ--",  # 2021 update (PERMENHUB 201)
        "https://ppid.dephub.go.id/ppid/file/Undangan_UU_No._17_Tahun_2008_Tentang_Pelayaran.pdf" # Shipping Act
        "https://jdih.esdm.go.id/",  # Ministry of Energy & Mineral Resources
        "https://maritim.go.id/",   # Coordinating Ministry for Maritime Affairs
        "https://peraturan.go.id/", # National Regulations Portal
    ],
    "Regional": [
        "https://asean.org/",             # ASEAN Secretariat
        "https://aseanenergy.org/",       # ASEAN Centre for Energy
        "https://www.unescap.org/",       # UN ESCAP (regional transport/energy)
        "https://www.oecd.org/environment/",  # OECD Environmental Policy
        "https://asean.org/book/asean-ports-sustainability-network-apsn-green-port-award-system-gpas/",
        "https://asean.org/wp-content/uploads/2020/12/ASEAN-Sustainable-Transport-Report-2020.pdf",
        "https://www.unescap.org/sites/default/files/ESCAP%20Maritime%20Transport%20Policy%20Framework.pdf",
    ],
    'Global': [
        'https://www.imo.org/en/OurWork/Environment/Pages/IMO-Strategy-on-reduction-of-GHG-emissions-from-ships.aspx',
        'https://www.apecpsn.org/apsn/u/cms/www/202308/14135311hj71.pdf'
        "https://imo.org/",               # International Maritime Organization
        "https://unctad.org/",            # UNCTAD (maritime transport)
        "https://www.worldbank.org/",     # World Bank (transport & climate)
        "https://www.iea.org/",           # International Energy Agency
        "https://unctad.org/system/files/official-document/dtltlb2019d2_en.pdf",  # Review of Maritime Transport 2019
        "https://unctad.org/system/files/official-document/rmt2023_en.pdf",
        "https://www.imo.org/en/OurWork/Environment/Pages/Default.aspx",  # IMO Environment
        "https://www.imo.org/en/MediaCentre/HotTopics/Pages/Cutting-GHG-emissions.aspx"
    ]
}


# ----------------------------
# Pillar keywords (heuristic evidence lists)
# Expanded: English, Bahasa Indonesia, Bahasa Melayu, Tagalog
# ----------------------------
pillars = {
    'standards': [
        # English + intl
        r'EE?XI', r'CI[Ii]', r'MARPOL', r'sulfur', r'low-?sulphur', r'EEXI', r'Emission[\s-]?control',
        # Bahasa Indonesia / Melayu
        r'PM 29', r'Peraturan Menteri Perhubungan', r'standar', r'peraturan', r'ketentuan',
        # Tagalog
        r'pamantayan', r'patakaran', r'alituntunin',
        # Thai
        r'มาตรฐาน', r'ข้อบังคับ', r'นโยบาย', r'กฎระเบียบ'
    ],
    'economic_instruments': [
        # English
        r'rebate', r'incentiv', r'grant', r'port dues', r'fee', r'levy', r'carbon price', r'green dues', r'subsid',
        # Bahasa Indonesia / Melayu
        r'insentif', r'subsidi', r'cukai', r'pengurangan bayaran', r'potongan', r'penghutang',
        # Tagalog
        r'insentibo', r'subsidiya', r'buwis', r'singil',
        # Thai
        r'แรงจูงใจ', r'เงินอุดหนุน', r'ภาษี', r'ค่าธรรมเนียม', r'ส่วนลด'
    ],
    'infrastructure': [
        # English
        r'OPS', r'shore[- ]power', r'bunk(er|ering)', r'alternative fuel', r'methanol', r'ammonia', r'LNG', r'liquefied natural gas', r'e-bunker', r'digital bunker',
        # Bahasa Indonesia / Melayu
        r'prasarana', r'infrastruktur hijau', r'elektrifikasi dermaga', r'tenaga darat', r'bahan bakar alternatif', r'gas cecair', r'biofuel', r'simpanan bahan bakar',
        # Tagalog
        r'impraestruktura', r'berde na pantalan', r'kuryente sa pantalan', r'alternatibong gasolina', r'imbakan ng gasolina',
        # Thai
        r'โครงสร้างพื้นฐาน', r'ท่าเรือสีเขียว', r'ไฟฟ้าฝั่ง', r'เชื้อเพลิงทางเลือก', r'ก๊าซธรรมชาติเหลว', r'เชื้อเพลิงชีวภาพ', r'สถานีเชื้อเพลิง'
    ],
    'MRV_transparency': [
        # English
        r'MRV', r'report', r'e-?bunk', r'bunker delivery note', r'electronic bunker', r'eBDN', r'emission reporting', r'transparen',
        # Bahasa Indonesia / Melayu
        r'laporan', r'pengukuran', r'pemantauan', r'pelaporan emisi', r'ketelusan', r'transparansi',
        # Tagalog
        r'ulat', r'pagsusuri', r'pagsusubaybay', r'pag-uulat ng emisyon', r'transparent',
        # Thai
        r'รายงาน', r'การตรวจสอบ', r'การติดตาม', r'การรายงานการปล่อย', r'ความโปร่งใส'
    ],
    'recognition': [
        # English
        r'GPAS', r'Green Port Award', r'ISO 14001', r'award', r'recognit', r'best practice',
        # Bahasa Indonesia / Melayu
        r'penghargaan', r'rekognisi', r'anugerah', r'praktik terbaik',
        # Tagalog
        r'parangal', r'pagkilala', r'pinakamahusay na gawi',
        # Thai
        r'รางวัล', r'การยอมรับ', r'แนวปฏิบัติที่ดีที่สุด'
    ],
   'Renewable': [
        # English
        r"\brenewable\b", r"solar", r"wind", r"hydro", r"geothermal", r"biomass",
        # Bahasa Indonesia
        r"energi terbarukan", r"tenaga surya", r"angin", r"air", r"panas bumi", r"biomassa",
        # Bahasa Melayu
        r"tenaga boleh baharu", r"tenaga suria", r"angin", r"hidro", r"geoterma", r"biomassa",
        # Tagalog
        r"renewable", r"solar", r"hangin", r"hidro", r"biomassa",
        # Thai
        r"พลังงานหมุนเวียน", r"พลังงานทดแทน", r"พลังงานแสงอาทิตย์", r"พลังงานลม", r"พลังน้ำ", r"พลังงานความร้อนใต้พิภพ", r"ชีวมวล"
    ],
    'Emission_Carbon': [
        # English
        r"emission", r"carbon", r"greenhouse", r"CO2",
        # Bahasa Indonesia
        r"emisi", r"karbon", r"gas rumah kaca", r"CO2",
        # Bahasa Melayu
        r"pelepasan", r"karbon", r"gas rumah hijau",
        # Tagalog
        r"emisyon", r"carbon", r"greenhouse gas",
        # Thai
        r"การปล่อยก๊าซ", r"คาร์บอน", r"ก๊าซเรือนกระจก", r"การปล่อย CO2"
    ],
    'Efficiency': [
        # English
        r"efficiency", r"energy saving", r"conservation",
        # Bahasa Indonesia
        r"efisiensi", r"hemat energi", r"konservasi energi",
        # Bahasa Melayu
        r"kecekapan tenaga", r"penjimatan tenaga", r"pemuliharaan tenaga",
        # Tagalog
        r"episyente", r"pagtitipid ng enerhiya",
        # Thai
        r"ประสิทธิภาพพลังงาน", r"การใช้พลังงานอย่างมีประสิทธิภาพ", r"การประหยัดพลังงาน"
    ],
    'Waste_Pollution': [
        # English
        r"waste", r"pollution", r"plastic", r"garbage", r"landfill",
        # Bahasa Indonesia
        r"limbah", r"sampah", r"polusi", r"TPA",
        # Bahasa Melayu
        r"sisa", r"sampah", r"pencemaran",
        # Tagalog
        r"basura", r"polusyon", r"plastik",
        # Thai
        r"ของเสีย", r"ขยะ", r"มลพิษ", r"กาก", r"หลุมฝังกลบ"
    ],
    'Policy_Law': [
        # English
        r"policy", r"law", r"regulation", r"act", r"agreement",
        # Bahasa Indonesia
        r"kebijakan", r"undang", r"peraturan", r"UU", r"kesepakatan",
        # Bahasa Melayu
        r"dasar", r"undang-undang", r"peraturan",
        # Tagalog
        r"patakaran", r"batas", r"regulasyon",
        # Thai
        r"นโยบาย", r"กฎหมาย", r"ข้อบังคับ", r"พระราชบัญญัติ", r"ข้อตกลง"
    ]
}

# normalize regex
for k in pillars:
    pillars[k] = [re.compile(pat, flags=re.I) for pat in pillars[k]]


# ----------------------------
# Helper: fetch and extract text from URL (HTML or PDF)
# ----------------------------

def fetch_text(url, timeout=20):
    try:
        r = requests.get(url, timeout=timeout)
        ct = r.headers.get('Content-Type','')
        if 'application/pdf' in ct or url.lower().endswith('.pdf'):
            # parse PDF
            with pdfplumber.open(io.BytesIO(r.content)) as pdf:
                pages = [p.extract_text() or '' for p in pdf.pages]
            text = '\n'.join(pages)
        else:
            soup = BeautifulSoup(r.text, 'html.parser')
            # remove scripts/styles
            for s in soup(['script','style','noscript']):
                s.decompose()
            text = soup.get_text(separator=' ', strip=True)
        return text
    except Exception as e:
        print(f"Failed to fetch {url}: {e}")
        return ''

# ----------------------------
# Crawl sources and build evidence counts per country
# ----------------------------
countries = ['Indonesia','Malaysia','Thailand','Singapore','Philippines']
years = list(range(2018, 2025))  # 2018..2024

# store raw texts (for review) and hits
raw_texts = {}
evidence = {c: {y: {p:0 for p in pillars} for y in years} for c in countries}

print('Fetching source pages (this may take a while)...')
for country, url_list in sources.items():
    for url in tqdm(url_list, desc=country):
        text = fetch_text(url)
        raw_texts[url] = text
        # naive time detection: find years mentioned in text; if none, assume applies to all recent years
        found_years = set(re.findall(r'20(1[8-9]|2[0-4])', text))
        found_years = {int('20'+y) for y in found_years}
        if not found_years:
            found_years = set(years)
        # match pillar keywords
        for p_name, patterns in pillars.items():
            matches = 0
            for pat in patterns:
                if pat.search(text):
                    matches += 1
            # assign evidence counts to found_years
            for y in found_years:
                if country in countries:
                    evidence[country][y][p_name] += matches
                else:
                    # regional/global -> add to all countries as weak evidence
                    for c in countries:
                        evidence[c][y][p_name] += max(0, matches-0)  # leave as is

# ----------------------------
# Map evidence counts -> ordinal score (0/1/2)
# Heuristic: 0 if 0 matches, 1 if 1-2 unique pattern groups, 2 if >=3
# ----------------------------

def map_to_ordinal(count):
    if count <= 0:
        return 0
    elif count <= 2:
        return 1
    else:
        return 2

rows = []
for c in countries:
    for y in years:
        row = {'country':c, 'year':y}
        for p in pillars:
            row[p] = map_to_ordinal(evidence[c][y][p])
        rows.append(row)

df = pd.DataFrame(rows)

# ----------------------------
# Compute GPRSI
# ----------------------------
component_cols = list(pillars.keys())
df['GPRSI_raw_mean'] = df[component_cols].mean(axis=1)
df['GPRSI_0_100'] = df['GPRSI_raw_mean'] * 50

# Min-max across sample per year-grouping optional
scaler = MinMaxScaler(feature_range=(0,100))
df['sum_score'] = df[component_cols].sum(axis=1)
df['GPRSI_minmax'] = scaler.fit_transform(df[['sum_score']])

# Save CSV for review
out_csv = 'gprsi_country_year_2018_2024.csv'
df.to_csv(out_csv, index=False)
print('\nSaved GPRSI CSV to', out_csv)

# ----------------------------
# Visualization: trend lines for each country
# ----------------------------
plt.figure(figsize=(10,5))
for c in countries:
    sub = df[df['country']==c]
    plt.plot(sub['year'], sub['GPRSI_0_100'], marker='o', label=c)
plt.title('GPRSI (0-100) by country — 2018-2024 (heuristic scrape)')
plt.xlabel('Year')
plt.ylabel('GPRSI (0-100)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Component radar-style for the latest year (2024 if present)
latest_year = max(years)
latest_df = df[df['year']==latest_year]
if latest_df.empty:
    latest_year = max(df['year'])
    latest_df = df[df['year']==latest_year]

# Parallel coordinates (normalized components)
plot_df = latest_df[['country']+component_cols].copy()
for ccol in component_cols:
    plot_df[ccol] = plot_df[ccol]/2.0
plt.figure(figsize=(10,5))
from pandas.plotting import parallel_coordinates
parallel_coordinates(plot_df, 'country')
plt.title(f'Normalized components comparison (0-1) — {latest_year}')
plt.ylabel('Normalized score (0-1)')
plt.tight_layout()
plt.show()

# ----------------------------
# Quick automated insights (text)
# ----------------------------
insights = []
# latest snapshot ranking
snapshot = latest_df.sort_values('GPRSI_0_100', ascending=False)
top = snapshot.iloc[0]
insights.append(f"Top performer in {latest_year} by heuristic GPRSI: {top['country']} ({top['GPRSI_0_100']:.1f}/100)")
insights.append(f"Mean GPRSI in {latest_year}: {snapshot['GPRSI_0_100'].mean():.1f}/100")
insights.append('Component variation: view the parallel coordinates plot for pillar-level differences.')

print('\n=== Automated Insights ===')
for s in insights:
    print('-', s)

# ----------------------------
# Save raw texts for manual review
# ----------------------------
import json
with open('raw_texts.json','w', encoding='utf-8') as f:
    json.dump({k:v for k,v in raw_texts.items()}, f)
print('Saved raw texts to raw_texts.json')

# End of notebook
print('\nNotebook run complete. PLEASE review raw_texts.json and the CSV to manually validate codings.')
```

```python
# -*- coding: utf-8 -*-
# =============================================================
# Colab-ready notebook script (paste into a Colab cell)
# Title: Scrape policy pages (2018–2024) for Green/Smart/Autonomous Port & Ship,
#        detect evidence, score pillars, and compute GPRSI+
# =============================================================
# What this does
# 1) Downloads curated policy pages / PDFs for Indonesia, Malaysia, Thailand,
#    Singapore, Philippines, plus regional/global items (as weak evidence)
# 2) Extracts text (HTML & PDF) with simple date-detection
# 3) Detects keyword evidence for policy pillars (incl. Smart Port & Autonomy)
# 4) Converts evidence into ordinal scores (0/1/2) per pillar per country-year (2018–2024)
# 5) Computes GPRSI+ and produces charts + CSV export
#
# NOTE: Heuristic, automated mapping. Always review raw_texts.json and CSV,
# then adjust codings manually for legal nuance and context.
# =============================================================

# ----------------------------
# Install dependencies (run once in Colab)
# ----------------------------
!pip install --quiet requests beautifulsoup4 pdfplumber pandas matplotlib scikit-learn tqdm lxml chardet

# ----------------------------
# Imports
# ----------------------------
import requests
from bs4 import BeautifulSoup
import pdfplumber
import io
import re
import pandas as pd
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from datetime import datetime
import json
import time

# ----------------------------
# Config
# ----------------------------
COUNTRIES = ['Indonesia','Malaysia','Thailand','Singapore','Philippines']
YEARS = list(range(2018, 2025))  # 2018..2024 inclusive
REGIONAL_WEIGHT = 0.5  # how strongly regional/global docs count vs national ones
TIMEOUT = 30
HEADERS = {
    'User-Agent': 'Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/123.0 Safari/537.36'
}

# ----------------------------
# Curated source URLs (fixed commas, add a few stability links)
# Add more URLs as you find them.
# ----------------------------
sources = {
    'Singapore': [
        'https://www.mpa.gov.sg/maritime-singapore/sustainability/maritime-singapore-green-initiative',
        'https://www.mpa.gov.sg/media-centre/details/port-marine-circular-no.-13-of-2024-revisions-to-the-maritime-singapore-green-initiative-%28msgi%29---integration-of-green-port-programme-%28gpp%29-into-the-enhanced-green-ship-programme-%28gsp%29-for-ocean-going-vessels',
        'https://www.mpa.gov.sg/media-centre/details/port-marine-circular-no.-12-of-2024-implementation-of-singapore-standard-ss-648--2024---code-of-practice-for-bunker-mass-flow-metering',
        'https://www.reuters.com/world/asia-pacific/singapore-mandate-digital-bunker-notes-starting-april-2025-2024-10-09/',
        'https://www.mpa.gov.sg/web/portal/home/mpa-maritime-singapore/green-initiatives/green-port-programme',
        'https://www.mpa.gov.sg/docs/mpaorg/pdf/MPA_Annual_Report_2023.pdf',
        'https://sustainabledevelopment.un.org/content/documents/commitments/commitment_1537.pdf',
        'https://www.mpa.gov.sg/',
        'https://www.nea.gov.sg/',
        'https://www.mti.gov.sg/'
    ],
    'Malaysia': [
        'https://www.pka.gov.my/phocadownload/policy/PORT%20KLANG%20GREEN%20POLICY.pdf',
        'https://www.transportevents.com/presentations/Manila2022/Port_Klang_Authority.pdf',
        'https://www.mot.gov.my/en/maritime/green-port',
        'https://www.pka.gov.my/index.php/en/green-initiatives',
        'https://www.marine.gov.my/jlm/resources/user_1/Misc/pdf/MSM/Malaysia_Shipping_Masterplan_2017-2022.pdf',
        'https://www.doe.gov.my/',
        'https://www.mot.gov.my/',
        'https://www.mea.gov.my/'
    ],
    'Philippines': [
        'https://www.ppa.com.ph/content/Supporting%20Documents/Item%201.c_PPA%20AO%20No.%2005-2018%20%20The%20Port%20Environmental%20Policy%20%28PEP%29_0.pdf',
        'https://www.scribd.com/document/379966171/Philippine-Ports-Authority-AO-No-05-2018-Port-Environmental-Policy',
        'https://marina.gov.ph/wp-content/uploads/2022/01/MARINA-Green-Port-Initiatives.pdf',
        'https://www.marina.gov.ph/reports/',
        'https://denr.gov.ph/images/DENR_Digital_Annual_Report_2022.pdf',
        'https://www.denr.gov.ph/',
        'https://www.ppa.com.ph/',
        'https://www.dti.gov.ph/'
    ],
    'Thailand': [
        'https://mddb.apec.org/Documents/2024/TPTWG/MEG-TM1/24_tptwg_meg_tm1_007.pdf',
        'https://pat.marketing/en/news/green-port/',
        'https://www.port.co.th/eng/sustainability',
        'https://www.md.go.th/statics/service/en/1147',
        'https://www.oie.go.th/sites/default/files/attachments/industrial_environment/Green%20Industry%20Handbook%202020.pdf',
        'https://www.port.co.th/',
        'https://www.onep.go.th/',
        'https://www.mot.go.th/'
    ],
    'Indonesia': [
        'https://peraturan.bpk.go.id/Details/103722/permenhub-no-29-tahun-2014',
        'https://budidjaja.law/2023/09/green-shipping-how-should-indonesia-react/',
        'https://www.dephub.go.id/post/read/peraturan-menteri-perhubungan-no-29-tahun-2018',
        'https://jdih.dephub.go.id/produk_hukum/view/VFVGNQ--',
        'https://ppid.dephub.go.id/ppid/file/Undangan_UU_No._17_Tahun_2008_Tentang_Pelayaran.pdf',
        'https://jdih.esdm.go.id/',
        'https://maritim.go.id/',
        'https://peraturan.go.id/'
    ],
    'Regional': [
        'https://asean.org/',
        'https://aseanenergy.org/',
        'https://www.unescap.org/',
        'https://www.oecd.org/environment/',
        'https://asean.org/book/asean-ports-sustainability-network-apsn-green-port-award-system-gpas/',
        'https://asean.org/wp-content/uploads/2020/12/ASEAN-Sustainable-Transport-Report-2020.pdf',
        'https://www.unescap.org/sites/default/files/ESCAP%20Maritime%20Transport%20Policy%20Framework.pdf'
    ],
    'Global': [
        'https://www.imo.org/en/OurWork/Environment/Pages/IMO-Strategy-on-reduction-of-GHG-emissions-from-ships.aspx',
        'https://www.apecpsn.org/apsn/u/cms/www/202308/14135311hj71.pdf',
        'https://imo.org/',
        'https://unctad.org/',
        'https://www.worldbank.org/',
        'https://www.iea.org/',
        'https://unctad.org/system/files/official-document/dtltlb2019d2_en.pdf',
        'https://unctad.org/system/files/official-document/rmt2023_en.pdf',
        'https://www.imo.org/en/OurWork/Environment/Pages/Default.aspx',
        'https://www.imo.org/en/MediaCentre/HotTopics/Pages/Cutting-GHG-emissions.aspx'
    ]
}

# ----------------------------
# Pillar keywords (expanded): Green/Sustainability + Smart Port + Autonomy
# English + Bahasa Indonesia + Bahasa Melayu + Tagalog + Thai
# ----------------------------
pillars = {
    # Core environmental & governance pillars
    'standards': [
        r'EE?XI', r'CI[Ii]', r'MARPOL', r'sulph?ur', r'Emission[\s-]?control', r'Annex VI',
        r'PM\s?29', r'Peraturan Menteri Perhubungan', r'standar', r'ketentuan', r'peraturan',
        r'pamantayan', r'patakaran', r'alituntunin',
        r'มาตรฐาน', r'ข้อบังคับ', r'นโยบาย', r'กฎระเบียบ'
    ],
    'economic_instruments': [
        r'rebate', r'incentiv', r'grant', r'port dues', r'fee', r'levy', r'carbon price', r'green dues', r'subsid',
        r'insentif', r'subsidi', r'cukai', r'pengurangan bayaran', r'potongan',
        r'insentibo', r'subsidiya', r'buwis', r'singil',
        r'แรงจูงใจ', r'เงินอุดหนุน', r'ภาษี', r'ค่าธรรมเนียม', r'ส่วนลด'
    ],
    'infrastructure': [
        r'OPS', r'shore[- ]power', r'bunk(er|ering)', r'alternative fuel', r'methanol', r'ammonia', r'LNG', r'liquefied natural gas', r'e-?bunker', r'digital bunker', r'biofuel',
        r'prasarana', r'infrastruktur hijau', r'elektrifikasi dermaga', r'tenaga darat', r'bahan bakar alternatif', r'gas cecair', r'simpanan bahan bakar',
        r'impraestruktura', r'berde na pantalan', r'kuryente sa pantalan', r'alternatibong gasolina',
        r'โครงสร้างพื้นฐาน', r'ท่าเรือสีเขียว', r'ไฟฟ้าฝั่ง', r'เชื้อเพลิงทางเลือก', r'ก๊าซธรรมชาติเหลว', r'เชื้อเพลิงชีวภาพ'
    ],
    'MRV_transparency': [
        r'\bMRV\b', r'report', r'e-?bunk', r'bunker delivery note', r'\beBDN\b', r'emission reporting', r'transparen', r'mass flow meter', r'SS[- ]648',
        r'laporan', r'pengukuran', r'pemantauan', r'pelaporan emisi', r'ketelusan', r'transparansi',
        r'ulat', r'pagsusuri', r'pagsusubaybay', r'pag-uulat ng emisyon',
        r'รายงาน', r'การตรวจสอบ', r'การติดตาม', r'ความโปร่งใส'
    ],
    'recognition': [
        r'\bGPAS\b', r'Green Port Award', r'ISO\s?14001', r'award', r'recognit', r'best practice',
        r'penghargaan', r'rekognisi', r'anugerah', r'praktik terbaik',
        r'parangal', r'pagkilala', r'แนวปฏิบัติที่ดีที่สุด', r'รางวัล'
    ],
    'Renewable': [
        r'\brenewable\b', r'solar', r'wind', r'hydro', r'geothermal', r'biomass',
        r'energi terbarukan', r'tenaga surya', r'angin', r'air', r'panas bumi', r'biomassa',
        r'tenaga boleh baharu', r'tenaga suria', r'hidro', r'geoterma',
        r'renewable', r'solar', r'hangin', r'hidro',
        r'พลังงานหมุนเวียน', r'พลังงานทดแทน', r'พลังงานแสงอาทิตย์', r'พลังงานลม', r'พลังน้ำ', r'ชีวมวล'
    ],
    'Emission_Carbon': [
        r'emission', r'carbon', r'greenhouse', r'\bCO2\b', r'GHG',
        r'emisi', r'karbon', r'gas rumah kaca',
        r'pelepasan', r'gas rumah hijau',
        r'emisyon', r'greenhouse gas',
        r'การปล่อยก๊าซ', r'คาร์บอน', r'ก๊าซเรือนกระจก'
    ],
    'Efficiency': [
        r'efficien', r'energy saving', r'conservation', r'optimization', r'eco[- ]efficien',
        r'efisiensi', r'hemat energi', r'konservasi energi', r'optimasi',
        r'kecekapan tenaga', r'penjimatan tenaga',
        r'episyente', r'pagtitipid ng enerhiya',
        r'ประสิทธิภาพพลังงาน', r'การประหยัดพลังงาน'
    ],
    'Waste_Pollution': [
        r'waste', r'pollution', r'plastic', r'garbage', r'landfill', r'wastewater', r'ballast water', r'scrubber',
        r'limbah', r'sampah', r'polusi', r'TPA', r'air limbah',
        r'sisa', r'pencemaran',
        r'basura', r'polusyon', r'plastik',
        r'ของเสีย', r'ขยะ', r'มลพิษ', r'หลุมฝังกลบ'
    ],
    'Policy_Law': [
        r'policy', r'law', r'regulation', r'act', r'agreement', r'decree', r'guideline',
        r'kebijakan', r'undang', r'peraturan', r'UU', r'keputusan menteri', r'pedoman',
        r'dasar', r'undang-undang', r'peraturan',
        r'patakaran', r'batas', r'regulasyon',
        r'นโยบาย', r'กฎหมาย', r'ข้อบังคับ', r'พระราชบัญญัติ'
    ],
    # New pillars focusing on Smart Port & Autonomy
    'Smart_Port': [
        r'\bsmart port\b', r'Port Community System', r'\bPCS\b', r'Single Window', r'National Single Window', r'NSW', r'TOS', r'IoT', r'5G', r'AI', r'machine learning', r'blockchain', r'digital twin', r'predictive', r'big data', r'analytics', r'cybersecurity',
        r'pelabuhan pintar', r'pelabuhan cerdas', r'sistem komunitas pelabuhan', r'jendela tunggal nasional', r'sistem operasi terminal', r'Internet of Things',
        r'pelabuhan pintar', r'pelabuhan bijak', r'komuniti pelabuhan', r'sistem operasi terminal',
        r'matalinong pantalan', r'smart port', r'single window', r'sistema ng komunidad ng pantalan',
        r'ท่าเรืออัจฉริยะ', r'ดิจิทัล', r'ดิจิไทเซชัน', r'บิ๊กดาต้า', r'ปัญญาประดิษฐ์'
    ],
    'Autonomous_Port': [
        r'automated terminal', r'automation', r'AGV', r'ASC', r'AQC', r'ASCs?', r'RTG automation', r'remote crane', r'remote[ -]?STS', r'OCR gate', r'无人|無人',
        r'terminal otomatis', r'otomatisasi', r'AGV', r'kendaraan terpandu otomatis', r'crane jarak jauh', r'gerbang OCR',
        r'terminal automasi', r'kren jauh',
        r'awtomatikong terminal', r'remote crane',
        r'ระบบท่าเรืออัตโนมัติ', r'ระบบอัตโนมัติ', r'ยานพาหนะไร้คนขับ'
    ],
    'Autonomous_Ship': [
        r'autonomous ship', r'autonomous vessel', r'\bMASS\b', r'remotely operated', r'unmanned surface vessel', r'\bUSV\b', r'autonomous navigation', r'COLREGs', r'e-Navigation', r'trial', r'testbed',
        r'kapal otonom', r'kapal tanpa awak', r'navigasi otonom', r'operasi jarak jauh',
        r'kapal autonomi', r'tanpa awak', r'navigasi autonomi',
        r'barko na awtonomo', r'sasakyang-dagat na walang sakay', r'paglalayag na awtonomo',
        r'เรืออัตโนมัติ', r'เดินเรืออัตโนมัติ', r'ควบคุมระยะไกล'
    ],
}

# compile regex patterns
for k in list(pillars.keys()):
    pillars[k] = [re.compile(pat, flags=re.I) for pat in pillars[k]]

# ----------------------------
# Helpers
# ----------------------------

def polite_get(url, timeout=TIMEOUT, headers=HEADERS, retries=2, sleep=1.0):
    for i in range(retries+1):
        try:
            r = requests.get(url, timeout=timeout, headers=headers)
            if r.status_code == 200:
                return r
        except Exception:
            pass
        time.sleep(sleep*(i+1))
    return None


def extract_text_from_response(r, url):
    ct = (r.headers.get('Content-Type','') or '').lower()
    if 'application/pdf' in ct or url.lower().endswith('.pdf'):
        try:
            with pdfplumber.open(io.BytesIO(r.content)) as pdf:
                pages = [p.extract_text() or '' for p in pdf.pages]
            return '\n'.join(pages)
        except Exception:
            return ''
    # HTML
    try:
        soup = BeautifulSoup(r.content, 'lxml')
    except Exception:
        soup = BeautifulSoup(r.text, 'html.parser')
    for s in soup(['script','style','noscript']):
        s.decompose()
    text = soup.get_text(separator=' ', strip=True)
    return text


def detect_years(text, url):
    """Try to detect publication years from text; fallback to regex over 2018–2024.
    If nothing found, return all YEARS.
    """
    years_found = set(int(y) for y in re.findall(r'\b(2018|2019|2020|2021|2022|2023|2024)\b', text))
    # Special-case: some URLs contain year
    url_years = set(int(y) for y in re.findall(r'\b(2018|2019|2020|2021|2022|2023|2024)\b', url))
    years_found |= url_years
    if not years_found:
        return set(YEARS)
    return years_found


# ----------------------------
# Crawl sources and build evidence counts per country
# ----------------------------
raw_texts = {}
evidence = {c: {y: {p:0.0 for p in pillars} for y in YEARS} for c in COUNTRIES}

print('Fetching source pages (this may take a while)...')
for bucket, url_list in sources.items():
    for url in tqdm(url_list, desc=f"{bucket}"):
        r = polite_get(url)
        if not r:
            print(f"Failed to fetch {url}")
            raw_texts[url] = ''
            continue
        text = extract_text_from_response(r, url)
        raw_texts[url] = text
        years = detect_years(text, url)
        # Count pillar matches
        for p_name, patterns in pillars.items():
            matches = 0
            for pat in patterns:
                if pat.search(text):
                    matches += 1
            if matches <= 0:
                continue
            # Assign to countries/years
            if bucket in COUNTRIES:
                for y in years:
                    evidence[bucket][y][p_name] += matches
            else:
                # regional/global -> apply as weak evidence across all countries
                for c in COUNTRIES:
                    for y in years:
                        evidence[c][y][p_name] += matches * REGIONAL_WEIGHT

# ----------------------------
# Map evidence counts -> ordinal score (0/1/2)
# Heuristic: 0 if 0 matches, 1 if 0< x <=2, 2 if x>2 (rounded for floats)
# ----------------------------

def map_to_ordinal(count):
    c = int(round(count))
    if c <= 0:
        return 0
    elif c <= 2:
        return 1
    else:
        return 2

rows = []
for c in COUNTRIES:
    for y in YEARS:
        row = {'country':c, 'year':y}
        for p in pillars:
            row[p] = map_to_ordinal(evidence[c][y][p])
        rows.append(row)

df = pd.DataFrame(rows)

# ----------------------------
# Compute GPRSI+
# ----------------------------
component_cols = list(pillars.keys())
df['GPRSI_raw_mean'] = df[component_cols].mean(axis=1)
df['GPRSI_0_100'] = df['GPRSI_raw_mean'] * 50  # since max per pillar=2 → mean∈[0,2] → *50 → [0,100]

# Optional: per-row sum + minmax
scaler = MinMaxScaler(feature_range=(0,100))
df['sum_score'] = df[component_cols].sum(axis=1)
df['GPRSI_minmax'] = scaler.fit_transform(df[['sum_score']])

# Save CSV for review
out_csv = 'gprsi_plus_country_year_2018_2024.csv'
df.to_csv(out_csv, index=False)
print('\nSaved GPRSI+ CSV to', out_csv)

# ----------------------------
# Visualization: trend lines for each country
# ----------------------------
plt.figure(figsize=(10,5))
for c in COUNTRIES:
    sub = df[df['country']==c]
    plt.plot(sub['year'], sub['GPRSI_0_100'], marker='o', label=c)
plt.title('GPRSI+ (0–100) by country — 2018–2024 (heuristic scrape)')
plt.xlabel('Year')
plt.ylabel('GPRSI+ (0–100)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Component comparison for the latest year
latest_year = max(YEARS)
latest_df = df[df['year']==latest_year]
if latest_df.empty:
    latest_year = int(df['year'].max())
    latest_df = df[df['year']==latest_year]

# Parallel coordinates (normalized components)
plot_df = latest_df[['country']+component_cols].copy()
for ccol in component_cols:
    plot_df[ccol] = plot_df[ccol]/2.0
from pandas.plotting import parallel_coordinates
plt.figure(figsize=(12,6))
parallel_coordinates(plot_df, 'country')
plt.title(f'Normalized components comparison (0–1) — {latest_year}')
plt.ylabel('Normalized score (0–1)')
plt.tight_layout()
plt.show()

# ----------------------------
# Quick automated insights (text)
# ----------------------------
insights = []
snapshot = latest_df.sort_values('GPRSI_0_100', ascending=False)
if not snapshot.empty:
    top = snapshot.iloc[0]
    insights.append(f"Top performer in {latest_year} by heuristic GPRSI+: {top['country']} ({top['GPRSI_0_100']:.1f}/100)")
    insights.append(f"Mean GPRSI+ in {latest_year}: {snapshot['GPRSI_0_100'].mean():.1f}/100")
    insights.append('Component variation: see the parallel-coordinates plot for pillar-level differences.')

print('\n=== Automated Insights ===')
for s in insights:
    print('-', s)

# ----------------------------
# Save raw texts for manual review
# ----------------------------
with open('raw_texts.json','w', encoding='utf-8') as f:
    json.dump(raw_texts, f, ensure_ascii=False)
print('Saved raw texts to raw_texts.json')

print('\nNotebook run complete. PLEASE review raw_texts.json and the CSV to manually validate codings.')
```

```python
# ----------------------------
# Follow-up cell: Contextual MAB baseline (LinUCB) using GPRSI CSV
# ----------------------------
# Requirements: scikit-learn, matplotlib, pandas, numpy (already in your Colab notebook)
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import math
import random
from collections import defaultdict

# ----------------------------
# Config / hyperparams
# ----------------------------
CSV_PATH = 'gprsi_country_year_2018_2024.csv'  # produced by your previous cell
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
random.seed(RANDOM_SEED)

# Bandit hyperparams
LINUCB_ALPHA = 1.0   # exploration parameter
N_ROUNDS = 2000      # number of bandit timesteps to simulate
VERBOSE = True

# ----------------------------
# Load data
# ----------------------------
try:
    df = pd.read_csv(CSV_PATH)
except FileNotFoundError:
    raise FileNotFoundError(f"Could not find {CSV_PATH}. Run the scraping cell first and ensure the CSV is saved.")

# Basic sanity: require columns
required_cols = {'country','year','GPRSI_0_100'}
if not required_cols.issubset(set(df.columns)):
    raise ValueError(f"CSV must contain columns {required_cols}. Found {df.columns.tolist()}")

# Sort & compute delta to next year (reward proxy)
df = df.sort_values(['country','year']).reset_index(drop=True)
df['GPRSI_next'] = df.groupby('country')['GPRSI_0_100'].shift(-1)
df = df.dropna(subset=['GPRSI_next']).copy()
df['delta'] = df['GPRSI_next'] - df['GPRSI_0_100']  # observed next-year change (can be negative)

# ----------------------------
# Define arms
# arms are policy levers (the pillars) — the bandit chooses which pillar to invest in
# Use the pillar column names from your dataset (subset of df.columns)
# ----------------------------
# detect pillar columns by presence in df
possible_pillars = ['standards','economic_instruments','infrastructure','MRV_transparency','recognition']
arms = [p for p in possible_pillars if p in df.columns]
if len(arms) == 0:
    raise ValueError("No pillar columns found in CSV. Ensure pillar columns (e.g., 'standards') exist.")

K = len(arms)
if VERBOSE:
    print(f"Detected arms (pillars): {arms}")

# ----------------------------
# Build context vectors
# Features: one-hot country, normalized year, current pillar scores, sum_score (if present)
# ----------------------------
# Country one-hot
countries = df['country'].unique().tolist()
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
country_ohe = ohe.fit_transform(df[['country']])

# Year normalized
year_scaler = StandardScaler()
year_norm = year_scaler.fit_transform(df[['year']])

# Pillar scores (as-is)
pillar_feats = df[arms].fillna(0).values.astype(float)

# optional sum_score if present
if 'sum_score' in df.columns:
    sum_score = df[['sum_score']].fillna(0).values.astype(float)
    X_base = np.hstack([country_ohe, year_norm, pillar_feats, sum_score])
else:
    X_base = np.hstack([country_ohe, year_norm, pillar_feats])

# Standardize contexts
ctx_scaler = StandardScaler()
X = ctx_scaler.fit_transform(X_base)

# Keep contexts and rewards
contexts = X  # each row is a context
rewards_observed = df['delta'].values  # used to train surrogate
n_examples = contexts.shape[0]
if VERBOSE:
    print(f"Prepared {n_examples} contexts (country-year observations). Context dimension = {contexts.shape[1]}")

# ----------------------------
# Train surrogate reward model
# We train a regressor that predicts delta (next-year GPRSI change) given (context + action)
# This model will act as the environment simulator: reward = surrogate.predict(context,action) + small_noise
# ----------------------------
# Create action one-hot features appended to context for training
action_ohe = np.eye(K)  # identity for K arms
# expand training data by making K copies for each example, one per possible action.
X_aug = np.repeat(contexts, K, axis=0)
actions_aug = np.tile(np.arange(K), n_examples)
action_features = action_ohe[actions_aug]
X_aug_full = np.hstack([X_aug, action_features])
y_aug = np.repeat(rewards_observed, K)  # naive: we use same observed delta for all hypothetical actions
# NOTE: because we don't have observed rewards per action, this surrogate is limited.
# To make the surrogate responsive to actions, we add a small synthetic signal:
# We'll inject a weak signal that actions aligned with pillar presence produce slightly higher reward.
# Construct a synthetic action effect based on current pillar score for that arm:
action_effect = np.zeros_like(y_aug)
for i in range(n_examples):
    base_idx = i*K
    for a in range(K):
        # use pillar score in original data for this arm as a proxy: higher existing pillar -> higher marginality? it's heuristic
        pillar_score = pillar_feats[i, a] if a < pillar_feats.shape[1] else 0.0
        action_effect[base_idx + a] = 0.2 * pillar_score  # small boost proportional to pillar_presence

y_aug = y_aug + action_effect

# Train regressor
rf = RandomForestRegressor(n_estimators=200, random_state=RANDOM_SEED, max_depth=6)
rf.fit(X_aug_full, y_aug)
if VERBOSE:
    print("Trained surrogate RandomForestRegressor (env simulator).")

# environment function: given context vector (1d) and chosen arm index -> predicted reward (float)
def env_reward_predict(context_vec, arm_idx, noise_std=0.05):
    a_onehot = np.zeros(K)
    a_onehot[arm_idx] = 1.0
    x_full = np.concatenate([context_vec, a_onehot]).reshape(1, -1)
    pred = rf.predict(x_full)[0]
    # small gaussian noise to make simulation stochastic
    return pred + np.random.normal(scale=noise_std)

# ----------------------------
# LinUCB implementation (per-arm linear models)
# Classic LinUCB (A_a = Dxd, b_a = Dx1) where context is d-dim vector
# ----------------------------
d = contexts.shape[1]
# initialize A and b for each arm
A = [np.eye(d) for _ in range(K)]
b = [np.zeros(d) for _ in range(K)]
theta = [np.zeros(d) for _ in range(K)]

def lincb_select_arm(context_vec, alpha=LINUCB_ALPHA):
    p_vals = np.zeros(K)
    for a in range(K):
        A_inv = np.linalg.inv(A[a])
        theta[a] = A_inv.dot(b[a])
        mean = theta[a].dot(context_vec)
        sigma = np.sqrt(context_vec.dot(A_inv).dot(context_vec))
        p_vals[a] = mean + alpha * sigma
    # choose arm with highest UCB
    return int(np.argmax(p_vals)), p_vals

# ----------------------------
# Simulation loop
# We'll sample contexts randomly from our observed contexts and simulate decisions for N_ROUNDS steps.
# Compare LinUCB to Random baseline and to an 'oracle' that always picks best arm per surrogate.
# ----------------------------
cumulative_reward_lincb = []
cumulative_reward_random = []
cumulative_reward_oracle = []

total_reward_lincb = 0.0
total_reward_random = 0.0
total_reward_oracle = 0.0

# track chosen counts
counts_lincb = np.zeros(K, dtype=int)
counts_random = np.zeros(K, dtype=int)

# precompute oracle choices (which arm gives highest predicted reward per context)
oracle_choice = []
for i in range(n_examples):
    ctx = contexts[i]
    preds = [env_reward_predict(ctx, a, noise_std=0.0) for a in range(K)]
    oracle_choice.append(int(np.argmax(preds)))

# run rounds
for t in range(N_ROUNDS):
    # sample a random context index (simulate encountering a country-year)
    idx = np.random.randint(0, n_examples)
    ctx = contexts[idx]

    # LinUCB selection
    arm_l, pvals = lincb_select_arm(ctx, alpha=LINUCB_ALPHA)
    r_l = env_reward_predict(ctx, arm_l)
    # update LinUCB parameters for chosen arm
    A[arm_l] += np.outer(ctx, ctx)
    b[arm_l] += r_l * ctx
    total_reward_lincb += r_l
    counts_lincb[arm_l] += 1
    cumulative_reward_lincb.append(total_reward_lincb)

    # Random baseline
    arm_r = np.random.randint(0, K)
    r_r = env_reward_predict(ctx, arm_r)
    total_reward_random += r_r
    counts_random[arm_r] += 1
    cumulative_reward_random.append(total_reward_random)

    # Oracle (cheating; uses surrogate to pick best)
    arm_o = oracle_choice[idx]
    r_o = env_reward_predict(ctx, arm_o)
    total_reward_oracle += r_o
    cumulative_reward_oracle.append(total_reward_oracle)

# ----------------------------
# Plot cumulative rewards
# ----------------------------
plt.figure(figsize=(10,6))
plt.plot(cumulative_reward_lincb, label=f"LinUCB (alpha={LINUCB_ALPHA})")
plt.plot(cumulative_reward_random, label="Random")
plt.plot(cumulative_reward_oracle, label="Oracle (surrogate best)")
plt.xlabel("Round")
plt.ylabel("Cumulative reward (simulated Δ GPRSI_0_100)")
plt.title("Contextual Bandit Baselines — cumulative reward")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# ----------------------------
# Summary statistics
# ----------------------------
print("\n--- Summary ---")
print(f"Rounds simulated: {N_ROUNDS}")
print(f"LinUCB total reward: {total_reward_lincb:.3f}")
print(f"Random total reward: {total_reward_random:.3f}")
print(f"Oracle total reward: {total_reward_oracle:.3f}")
print("\nLinUCB chosen counts per arm:")
for i,a in enumerate(arms):
    print(f"  {a}: {counts_lincb[i]}")
print("\nRandom chosen counts per arm:")
for i,a in enumerate(arms):
    print(f"  {a}: {counts_random[i]}")

# Optionally: save results to CSV
out_hist = pd.DataFrame({
    'round': np.arange(1, N_ROUNDS+1),
    'cum_reward_lincb': cumulative_reward_lincb,
    'cum_reward_random': cumulative_reward_random,
    'cum_reward_oracle': cumulative_reward_oracle
})
out_hist.to_csv('mab_simulation_history.csv', index=False)
print("\nSaved simulation history to mab_simulation_history.csv")
```

```python
# Colab-ready notebook: Scrape policy pages (2018-2024) and build GPRSI
# Instructions: Run this notebook in Google Colab (internet-enabled). It will:
# 1) Download a curated set of policy pages / PDFs for Indonesia, Malaysia, Thailand, Singapore, Philippines, and regional/global items
# 2) Extract text and detect keyword evidence for policy pillars
# 3) Convert evidence into ordinal scores (0/1/2) per pillar per country-year (2018-2024)
# 4) Compute GPRSI and produce charts + CSV export

# NOTE: This is an automated, heuristic approach. You should review and adjust codings manually
# after running (especially for nuanced legal texts). The heuristic maps keyword hits -> ordinal scores.

# ----------------------------
# Install dependencies (run once)
# ----------------------------
!pip install --quiet requests beautifulsoup4 pdfplumber pandas matplotlib scikit-learn tqdm

# ----------------------------
# Imports
# ----------------------------
import requests
from bs4 import BeautifulSoup
import pdfplumber
import io
import re
import pandas as pd
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

# ----------------------------
# Curated source URLs (evidence seeds)
# Add more URLs as you find them.
# ----------------------------
sources = {
    'Singapore': [
        'https://www.mpa.gov.sg/maritime-singapore/sustainability/maritime-singapore-green-initiative',
        'https://www.mpa.gov.sg/media-centre/details/port-marine-circular-no.-13-of-2024-revisions-to-the-maritime-singapore-green-initiative-%28msgi%29---integration-of-green-port-programme-%28gpp%29-into-the-enhanced-green-ship-programme-%28gsp%29-for-ocean-going-vessels',
        'https://www.mpa.gov.sg/media-centre/details/port-marine-circular-no.-12-of-2024-implementation-of-singapore-standard-ss-648--2024---code-of-practice-for-bunker-mass-flow-metering',
        'https://www.reuters.com/world/asia-pacific/singapore-mandate-digital-bunker-notes-starting-april-2025-2024-10-09/',
        "https://www.mpa.gov.sg/web/portal/home/mpa-maritime-singapore/green-initiatives/green-port-programme",
        "https://www.mpa.gov.sg/docs/mpaorg/pdf/MPA_Annual_Report_2023.pdf",
        "https://sustainabledevelopment.un.org/content/documents/commitments/commitment_1537.pdf"
        "https://www.mpa.gov.sg/",   # Maritime and Port Authority
        "https://www.nea.gov.sg/",   # National Environment Agency
        "https://www.mti.gov.sg/",   # Ministry of Trade and Industry
    ],
    'Malaysia': [
        'https://www.pka.gov.my/phocadownload/policy/PORT%20KLANG%20GREEN%20POLICY.pdf',
        'https://www.transportevents.com/presentations/Manila2022/Port_Klang_Authority.pdf',
         "https://www.mot.gov.my/en/maritime/green-port",
        "https://www.pka.gov.my/index.php/en/green-initiatives",  # Port Klang green program
        "https://www.marine.gov.my/jlm/resources/user_1/Misc/pdf/MSM/Malaysia_Shipping_Masterplan_2017-2022.pdf"
        "https://www.doe.gov.my/",   # Department of Environment
        "https://www.mot.gov.my/",   # Ministry of Transport (green shipping policy)
        "https://www.mea.gov.my/",   # Ministry of Economic Affairs
    ],
    'Philippines': [
        'https://www.ppa.com.ph/content/Supporting%20Documents/Item%201.c_PPA%20AO%20No.%2005-2018%20%20The%20Port%20Environmental%20Policy%20%28PEP%29_0.pdf',
        'https://www.scribd.com/document/379966171/Philippine-Ports-Authority-AO-No-05-2018-Port-Environmental-Policy'
        "https://marina.gov.ph/wp-content/uploads/2022/01/MARINA-Green-Port-Initiatives.pdf",
        "https://www.marina.gov.ph/reports/",
        "https://denr.gov.ph/images/DENR_Digital_Annual_Report_2022.pdf"  # DENR annual report
        "https://www.denr.gov.ph/",  # Department of Environment and Natural Resources
        "https://www.ppa.com.ph/",   # Philippine Ports Authority
        "https://www.dti.gov.ph/",   # Department of Trade and Industry
    ],
    'Thailand': [
        'https://mddb.apec.org/Documents/2024/TPTWG/MEG-TM1/24_tptwg_meg_tm1_007.pdf',
        'https://pat.marketing/en/news/green-port/',
        "https://www.port.co.th/eng/sustainability",
        "https://www.md.go.th/statics/service/en/1147",  # Marine Department – Green Port Policy
        "https://www.oie.go.th/sites/default/files/attachments/industrial_environment/Green%20Industry%20Handbook%202020.pdf"
        "https://www.port.co.th/",   # Port Authority of Thailand
        "https://www.onep.go.th/",   # Office of Natural Resources & Environmental Policy
        "https://www.mot.go.th/",    # Ministry of Transport
    ],
    'Indonesia': [
        'https://peraturan.bpk.go.id/Details/103722/permenhub-no-29-tahun-2014',
        'https://budidjaja.law/2023/09/green-shipping-how-should-indonesia-react/',
        "https://www.dephub.go.id/post/read/peraturan-menteri-perhubungan-no-29-tahun-2018",   # Green Port Guidelines
        "https://jdih.dephub.go.id/produk_hukum/view/VFVGNQ--",  # 2021 update (PERMENHUB 201)
        "https://ppid.dephub.go.id/ppid/file/Undangan_UU_No._17_Tahun_2008_Tentang_Pelayaran.pdf" # Shipping Act
        "https://jdih.esdm.go.id/",  # Ministry of Energy & Mineral Resources
        "https://maritim.go.id/",   # Coordinating Ministry for Maritime Affairs
        "https://peraturan.go.id/", # National Regulations Portal
    ],
    "Regional": [
        "https://asean.org/",             # ASEAN Secretariat
        "https://aseanenergy.org/",       # ASEAN Centre for Energy
        "https://www.unescap.org/",       # UN ESCAP (regional transport/energy)
        "https://www.oecd.org/environment/",  # OECD Environmental Policy
        "https://asean.org/book/asean-ports-sustainability-network-apsn-green-port-award-system-gpas/",
        "https://asean.org/wp-content/uploads/2020/12/ASEAN-Sustainable-Transport-Report-2020.pdf",
        "https://www.unescap.org/sites/default/files/ESCAP%20Maritime%20Transport%20Policy%20Framework.pdf",
    ],
    'Global': [
        'https://www.imo.org/en/OurWork/Environment/Pages/IMO-Strategy-on-reduction-of-GHG-emissions-from-ships.aspx',
        'https://www.apecpsn.org/apsn/u/cms/www/202308/14135311hj71.pdf'
        "https://imo.org/",               # International Maritime Organization
        "https://unctad.org/",            # UNCTAD (maritime transport)
        "https://www.worldbank.org/",     # World Bank (transport & climate)
        "https://www.iea.org/",           # International Energy Agency
        "https://unctad.org/system/files/official-document/dtltlb2019d2_en.pdf",  # Review of Maritime Transport 2019
        "https://unctad.org/system/files/official-document/rmt2023_en.pdf",
        "https://www.imo.org/en/OurWork/Environment/Pages/Default.aspx",  # IMO Environment
        "https://www.imo.org/en/MediaCentre/HotTopics/Pages/Cutting-GHG-emissions.aspx"
    ]
}


# ----------------------------
# Pillar keywords (heuristic evidence lists)
# Expanded: English, Bahasa Indonesia, Bahasa Melayu, Tagalog
# ----------------------------
pillars = {
    'standards': [
        # English + intl
        r'EE?XI', r'CI[Ii]', r'MARPOL', r'sulfur', r'low-?sulphur', r'EEXI', r'Emission[\s-]?control',
        # Bahasa Indonesia / Melayu
        r'PM 29', r'Peraturan Menteri Perhubungan', r'standar', r'peraturan', r'ketentuan',
        # Tagalog
        r'pamantayan', r'patakaran', r'alituntunin',
        # Thai
        r'มาตรฐาน', r'ข้อบังคับ', r'นโยบาย', r'กฎระเบียบ'
    ],
    'economic_instruments': [
        # English
        r'rebate', r'incentiv', r'grant', r'port dues', r'fee', r'levy', r'carbon price', r'green dues', r'subsid',
        # Bahasa Indonesia / Melayu
        r'insentif', r'subsidi', r'cukai', r'pengurangan bayaran', r'potongan', r'penghutang',
        # Tagalog
        r'insentibo', r'subsidiya', r'buwis', r'singil',
        # Thai
        r'แรงจูงใจ', r'เงินอุดหนุน', r'ภาษี', r'ค่าธรรมเนียม', r'ส่วนลด'
    ],
    'infrastructure': [
        # English
        r'OPS', r'shore[- ]power', r'bunk(er|ering)', r'alternative fuel', r'methanol', r'ammonia', r'LNG', r'liquefied natural gas', r'e-bunker', r'digital bunker',
        # Bahasa Indonesia / Melayu
        r'prasarana', r'infrastruktur hijau', r'elektrifikasi dermaga', r'tenaga darat', r'bahan bakar alternatif', r'gas cecair', r'biofuel', r'simpanan bahan bakar',
        # Tagalog
        r'impraestruktura', r'berde na pantalan', r'kuryente sa pantalan', r'alternatibong gasolina', r'imbakan ng gasolina',
        # Thai
        r'โครงสร้างพื้นฐาน', r'ท่าเรือสีเขียว', r'ไฟฟ้าฝั่ง', r'เชื้อเพลิงทางเลือก', r'ก๊าซธรรมชาติเหลว', r'เชื้อเพลิงชีวภาพ', r'สถานีเชื้อเพลิง'
    ],
    'MRV_transparency': [
        # English
        r'MRV', r'report', r'e-?bunk', r'bunker delivery note', r'electronic bunker', r'eBDN', r'emission reporting', r'transparen',
        # Bahasa Indonesia / Melayu
        r'laporan', r'pengukuran', r'pemantauan', r'pelaporan emisi', r'ketelusan', r'transparansi',
        # Tagalog
        r'ulat', r'pagsusuri', r'pagsusubaybay', r'pag-uulat ng emisyon', r'transparent',
        # Thai
        r'รายงาน', r'การตรวจสอบ', r'การติดตาม', r'การรายงานการปล่อย', r'ความโปร่งใส'
    ],
    'recognition': [
        # English
        r'GPAS', r'Green Port Award', r'ISO 14001', r'award', r'recognit', r'best practice',
        # Bahasa Indonesia / Melayu
        r'penghargaan', r'rekognisi', r'anugerah', r'praktik terbaik',
        # Tagalog
        r'parangal', r'pagkilala', r'pinakamahusay na gawi',
        # Thai
        r'รางวัล', r'การยอมรับ', r'แนวปฏิบัติที่ดีที่สุด'
    ],
   'Renewable': [
        # English
        r"\brenewable\b", r"solar", r"wind", r"hydro", r"geothermal", r"biomass",
        # Bahasa Indonesia
        r"energi terbarukan", r"tenaga surya", r"angin", r"air", r"panas bumi", r"biomassa",
        # Bahasa Melayu
        r"tenaga boleh baharu", r"tenaga suria", r"angin", r"hidro", r"geoterma", r"biomassa",
        # Tagalog
        r"renewable", r"solar", r"hangin", r"hidro", r"biomassa",
        # Thai
        r"พลังงานหมุนเวียน", r"พลังงานทดแทน", r"พลังงานแสงอาทิตย์", r"พลังงานลม", r"พลังน้ำ", r"พลังงานความร้อนใต้พิภพ", r"ชีวมวล"
    ],
    'Emission_Carbon': [
        # English
        r"emission", r"carbon", r"greenhouse", r"CO2",
        # Bahasa Indonesia
        r"emisi", r"karbon", r"gas rumah kaca", r"CO2",
        # Bahasa Melayu
        r"pelepasan", r"karbon", r"gas rumah hijau",
        # Tagalog
        r"emisyon", r"carbon", r"greenhouse gas",
        # Thai
        r"การปล่อยก๊าซ", r"คาร์บอน", r"ก๊าซเรือนกระจก", r"การปล่อย CO2"
    ],
    'Efficiency': [
        # English
        r"efficiency", r"energy saving", r"conservation",
        # Bahasa Indonesia
        r"efisiensi", r"hemat energi", r"konservasi energi",
        # Bahasa Melayu
        r"kecekapan tenaga", r"penjimatan tenaga", r"pemuliharaan tenaga",
        # Tagalog
        r"episyente", r"pagtitipid ng enerhiya",
        # Thai
        r"ประสิทธิภาพพลังงาน", r"การใช้พลังงานอย่างมีประสิทธิภาพ", r"การประหยัดพลังงาน"
    ],
    'Waste_Pollution': [
        # English
        r"waste", r"pollution", r"plastic", r"garbage", r"landfill",
        # Bahasa Indonesia
        r"limbah", r"sampah", r"polusi", r"TPA",
        # Bahasa Melayu
        r"sisa", r"sampah", r"pencemaran",
        # Tagalog
        r"basura", r"polusyon", r"plastik",
        # Thai
        r"ของเสีย", r"ขยะ", r"มลพิษ", r"กาก", r"หลุมฝังกลบ"
    ],
    'Policy_Law': [
        # English
        r"policy", r"law", r"regulation", r"act", r"agreement",
        # Bahasa Indonesia
        r"kebijakan", r"undang", r"peraturan", r"UU", r"kesepakatan",
        # Bahasa Melayu
        r"dasar", r"undang-undang", r"peraturan",
        # Tagalog
        r"patakaran", r"batas", r"regulasyon",
        # Thai
        r"นโยบาย", r"กฎหมาย", r"ข้อบังคับ", r"พระราชบัญญัติ", r"ข้อตกลง"
    ],
    'Green_Port_Tech': [
    r'green port', r'eco[- ]?port', r'smart port', r'digital port', r'ecoport',
    r'pelabuhan hijau', r'pelabuhan pintar', r'berdeng pantalan', r'matalinong pantalan',
    r'ท่าเรือสีเขียว', r'ท่าเรืออัจฉริยะ'
],
'Alt_Fuels_Tech': [
    r'LNG', r'liquefied natural gas', r'methanol', r'ammonia', r'hydrogen', r'H2',
    r'biofuel', r'biodiesel', r'e-fuel', r'e-methanol', r'e-ammonia',
    r'CCUS', r'carbon capture', r'fuel cell',
    r'bahan bakar alternatif', r'hidrogen', r'metanol', r'amonia',
    r'alternatibong gasolina', r'hidrogeno', r'метанол',
    r'เชื้อเพลิงทางเลือก', r'ไฮโดรเจน', r'เมทานอล', r'แอมโมเนีย'
]

}

# normalize regex
for k in pillars:
    pillars[k] = [re.compile(pat, flags=re.I) for pat in pillars[k]]


# ----------------------------
# Helper: fetch and extract text from URL (HTML or PDF)
# ----------------------------

def fetch_text(url, timeout=20):
    try:
        r = requests.get(url, timeout=timeout)
        ct = r.headers.get('Content-Type','')
        if 'application/pdf' in ct or url.lower().endswith('.pdf'):
            # parse PDF
            with pdfplumber.open(io.BytesIO(r.content)) as pdf:
                pages = [p.extract_text() or '' for p in pdf.pages]
            text = '\n'.join(pages)
        else:
            soup = BeautifulSoup(r.text, 'html.parser')
            # remove scripts/styles
            for s in soup(['script','style','noscript']):
                s.decompose()
            text = soup.get_text(separator=' ', strip=True)
        return text
    except Exception as e:
        print(f"Failed to fetch {url}: {e}")
        return ''

# ----------------------------
# Crawl sources and build evidence counts per country
# ----------------------------
countries = ['Indonesia','Malaysia','Thailand','Singapore','Philippines']
years = list(range(2018, 2025))  # 2018..2024

# store raw texts (for review) and hits
raw_texts = {}
evidence = {c: {y: {p:0 for p in pillars} for y in years} for c in countries}

print('Fetching source pages (this may take a while)...')
for country, url_list in sources.items():
    for url in tqdm(url_list, desc=country):
        text = fetch_text(url)
        raw_texts[url] = text
        # naive time detection: find years mentioned in text; if none, assume applies to all recent years
        found_years = set(re.findall(r'20(1[8-9]|2[0-4])', text))
        found_years = {int('20'+y) for y in found_years}
        if not found_years:
            found_years = set(years)
        # match pillar keywords
        for p_name, patterns in pillars.items():
            matches = 0
            for pat in patterns:
                if pat.search(text):
                    matches += 1
            # assign evidence counts to found_years
            for y in found_years:
                if country in countries:
                    evidence[country][y][p_name] += matches
                else:
                    # regional/global -> add to all countries as weak evidence
                    for c in countries:
                        evidence[c][y][p_name] += max(0, matches-0)  # leave as is

# ----------------------------
# Map evidence counts -> ordinal score (0/1/2)
# Heuristic: 0 if 0 matches, 1 if 1-2 unique pattern groups, 2 if >=3
# ----------------------------

def map_to_ordinal(count):
    if count <= 0:
        return 0
    elif count <= 2:
        return 1
    else:
        return 2

rows = []
for c in countries:
    for y in years:
        row = {'country':c, 'year':y}
        for p in pillars:
            row[p] = map_to_ordinal(evidence[c][y][p])
        rows.append(row)

df = pd.DataFrame(rows)

# ----------------------------
# Compute GPRSI
# ----------------------------
component_cols = list(pillars.keys())
df['GPRSI_raw_mean'] = df[component_cols].mean(axis=1)
df['GPRSI_0_100'] = df['GPRSI_raw_mean'] * 50

# Min-max across sample per year-grouping optional
scaler = MinMaxScaler(feature_range=(0,100))
df['sum_score'] = df[component_cols].sum(axis=1)
df['GPRSI_minmax'] = scaler.fit_transform(df[['sum_score']])

# Save CSV for review
out_csv = 'gprsi_country_year_2018_2024.csv'
df.to_csv(out_csv, index=False)
print('\nSaved GPRSI CSV to', out_csv)

# ----------------------------
# Visualization: trend lines for each country
# ----------------------------
plt.figure(figsize=(10,5))
for c in countries:
    sub = df[df['country']==c]
    plt.plot(sub['year'], sub['GPRSI_0_100'], marker='o', label=c)
plt.title('GPRSI (0-100) by country — 2018-2024 (heuristic scrape)')
plt.xlabel('Year')
plt.ylabel('GPRSI (0-100)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Component radar-style for the latest year (2024 if present)
latest_year = max(years)
latest_df = df[df['year']==latest_year]
if latest_df.empty:
    latest_year = max(df['year'])
    latest_df = df[df['year']==latest_year]

# Parallel coordinates (normalized components)
plot_df = latest_df[['country']+component_cols].copy()
for ccol in component_cols:
    plot_df[ccol] = plot_df[ccol]/2.0
plt.figure(figsize=(10,5))
from pandas.plotting import parallel_coordinates
parallel_coordinates(plot_df, 'country')
plt.title(f'Normalized components comparison (0-1) — {latest_year}')
plt.ylabel('Normalized score (0-1)')
plt.tight_layout()
plt.show()

# ----------------------------
# Quick automated insights (text)
# ----------------------------
insights = []
# latest snapshot ranking
snapshot = latest_df.sort_values('GPRSI_0_100', ascending=False)
top = snapshot.iloc[0]
insights.append(f"Top performer in {latest_year} by heuristic GPRSI: {top['country']} ({top['GPRSI_0_100']:.1f}/100)")
insights.append(f"Mean GPRSI in {latest_year}: {snapshot['GPRSI_0_100'].mean():.1f}/100")
insights.append('Component variation: view the parallel coordinates plot for pillar-level differences.')

print('\n=== Automated Insights ===')
for s in insights:
    print('-', s)

# ----------------------------
# Save raw texts for manual review
# ----------------------------
import json
with open('raw_texts.json','w', encoding='utf-8') as f:
    json.dump({k:v for k,v in raw_texts.items()}, f)
print('Saved raw texts to raw_texts.json')

# End of notebook
print('\nNotebook run complete. PLEASE review raw_texts.json and the CSV to manually validate codings.')
```

```python
# Colab-ready notebook: Scrape policy pages (2018-2024) and build GPRSI
# Instructions: Run this notebook in Google Colab (internet-enabled). It will:
# 1) Download a curated set of policy pages / PDFs for Indonesia, Malaysia, Thailand, Singapore, Philippines, and regional/global items
# 2) Extract text and detect keyword evidence for policy pillars
# 3) Convert evidence into ordinal scores (0/1/2) per pillar per country-year (2018-2024)
# 4) Compute GPRSI and produce charts + CSV export

# NOTE: This is an automated, heuristic approach. You should review and adjust codings manually
# after running (especially for nuanced legal texts). The heuristic maps keyword hits -> ordinal scores.

# ----------------------------
# Install dependencies (run once)
# ----------------------------
!pip install --quiet requests beautifulsoup4 pdfplumber pandas matplotlib scikit-learn tqdm

# ----------------------------
# Imports
# ----------------------------
import requests
from bs4 import BeautifulSoup
import pdfplumber
import io
import re
import pandas as pd
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

# ----------------------------
# Curated source URLs (evidence seeds)
# Add more URLs as you find them.
# ----------------------------
sources = {
   'Singapore': [
    "https://www.mpa.gov.sg/maritime-singapore/sustainability/maritime-singapore-green-initiative",
    "https://www.mpa.gov.sg/web/portal/home/mpa-maritime-singapore/green-initiatives/green-port-programme",
    "https://www.mpa.gov.sg/maritime-singapore/sustainability/maritime-singapore-decarbonisation-blueprint",
    "https://www.mpa.gov.sg/docs/mpalibraries/mpa-documents-files/sustainability-office/mpa-decarb-blueprint-2050a.pdf",
    "https://www.mot.gov.sg/what-we-do/green-transport/maritime-environment-responsibility/",
    "https://www.unescap.org/sites/default/d8files/event-documents/Sustainable_and_resilient_Pacific_Ports_Feb_2022_Final.pdf"
],
    'Malaysia': [
    "https://www.pka.gov.my/phocadownload/policy/PORT%20KLANG%20GREEN%20POLICY.pdf",
    "https://www.mot.gov.my/en/maritime",
    "https://www.mot.gov.my/en/Documents/Port%20Authorities%20Act%201963%20(act%20488)_2.pdf",
    "https://www.mot.gov.my/en/Documents/Ports%20(Privatization)%20Act%201990%20(act%20422).pdf",
    "https://www.investmalaysia.gov.my/media/m3eamxey/thirteenth-malaysia-plan-2026-2030.pdf",
    "https://www.unescap.org/sites/default/d8files/event-documents/TS2_Country%20presentation_Malaysia.pdf"
],

    'Philippines': [
    "https://www.ppa.com.ph/content/Supporting%20Documents/Item%201.c_PPA%20AO%20No.%2005-2018%20%20The%20Port%20Environmental%20Policy%20%28PEP%29_0.pdf",
    "https://www.ppa.com.ph/",
    "https://marina.gov.ph/wp-content/uploads/2024/04/MIDP-2028-as-of-29-April-2024.pdf",
    "https://www.denr.gov.ph/",
    "https://denr.gov.ph/news-events/green-shipping-coalition-roundtable-discussion/",
    "https://www.unescap.org/sites/default/d8files/event-documents/TS2_Country%20presentation_Philippines.pdf"
],
    'Thailand': [
    "https://pat.marketing/en/news/green-port/",
    "https://www.port.co.th/",
    "https://www.mot.go.th/",
    "https://www.onep.go.th/",
    "https://md.go.th/wp-content/uploads/2021/03/full-translated_Thailands-Overall-Maritime-Strategy-as-of-21-01-2023.pdf",
    "https://www.unescap.org/sites/default/d8files/event-documents/18-Thailand.pdf"
],
    'Indonesia': [
    "https://peraturan.bpk.go.id/Details/103722/permenhub-no-29-tahun-2014",
    "https://peraturan.go.id/",
    "https://maritim.go.id/",
    "https://jdih.esdm.go.id/",
    "https://perpustakaan.bappenas.go.id/e-library/file_upload/koleksi/migrasi-data-publikasi/file/Unit_Kerja/Dir%20Industri%2C%20Ekonomi%20dan%20Kreatif/Dummy_Indonesia%20Blue%20Economy%20Roadmap_Ebook.pdf",
    "https://www.unescap.org/sites/default/files/6.Indonesia.pdf"
],
   'Regional_Global_Reference': [
    "https://asean.org/",
    "https://asean.org/wp-content/uploads/2023/08/AMO-1.pdf",
    "https://www.unescap.org/",
    "https://www.imo.org/en/OurWork/Environment/Pages/IMO-Strategy-on-reduction-of-GHG-emissions-from-ships.aspx",
    "https://unctad.org/system/files/official-document/rmt2023_en.pdf",
    "https://www.oecd.org/environment/"
]
}


# ----------------------------
# Pillar keywords (heuristic evidence lists)
# Expanded: English, Bahasa Indonesia, Bahasa Melayu, Tagalog
# ----------------------------
pillars = {
    'standards': [
        # English + intl
        r'EE?XI', r'CI[Ii]', r'MARPOL', r'sulfur', r'low-?sulphur', r'EEXI', r'Emission[\s-]?control',
        # Bahasa Indonesia / Melayu
        r'PM 29', r'Peraturan Menteri Perhubungan', r'standar', r'peraturan', r'ketentuan',
        # Tagalog
        r'pamantayan', r'patakaran', r'alituntunin',
        # Thai
        r'มาตรฐาน', r'ข้อบังคับ', r'นโยบาย', r'กฎระเบียบ'
    ],
    'economic_instruments': [
        # English
        r'rebate', r'incentiv', r'grant', r'port dues', r'fee', r'levy', r'carbon price', r'green dues', r'subsid',
        # Bahasa Indonesia / Melayu
        r'insentif', r'subsidi', r'cukai', r'pengurangan bayaran', r'potongan', r'penghutang',
        # Tagalog
        r'insentibo', r'subsidiya', r'buwis', r'singil',
        # Thai
        r'แรงจูงใจ', r'เงินอุดหนุน', r'ภาษี', r'ค่าธรรมเนียม', r'ส่วนลด'
    ],
    'infrastructure': [
        # English
        r'OPS', r'shore[- ]power', r'bunk(er|ering)', r'alternative fuel', r'methanol', r'ammonia', r'LNG', r'liquefied natural gas', r'e-bunker', r'digital bunker',
        # Bahasa Indonesia / Melayu
        r'prasarana', r'infrastruktur hijau', r'elektrifikasi dermaga', r'tenaga darat', r'bahan bakar alternatif', r'gas cecair', r'biofuel', r'simpanan bahan bakar',
        # Tagalog
        r'impraestruktura', r'berde na pantalan', r'kuryente sa pantalan', r'alternatibong gasolina', r'imbakan ng gasolina',
        # Thai
        r'โครงสร้างพื้นฐาน', r'ท่าเรือสีเขียว', r'ไฟฟ้าฝั่ง', r'เชื้อเพลิงทางเลือก', r'ก๊าซธรรมชาติเหลว', r'เชื้อเพลิงชีวภาพ', r'สถานีเชื้อเพลิง'
    ],
    'MRV_transparency': [
        # English
        r'MRV', r'report', r'e-?bunk', r'bunker delivery note', r'electronic bunker', r'eBDN', r'emission reporting', r'transparen',
        # Bahasa Indonesia / Melayu
        r'laporan', r'pengukuran', r'pemantauan', r'pelaporan emisi', r'ketelusan', r'transparansi',
        # Tagalog
        r'ulat', r'pagsusuri', r'pagsusubaybay', r'pag-uulat ng emisyon', r'transparent',
        # Thai
        r'รายงาน', r'การตรวจสอบ', r'การติดตาม', r'การรายงานการปล่อย', r'ความโปร่งใส'
    ],
    'recognition': [
        # English
        r'GPAS', r'Green Port Award', r'ISO 14001', r'award', r'recognit', r'best practice',
        # Bahasa Indonesia / Melayu
        r'penghargaan', r'rekognisi', r'anugerah', r'praktik terbaik',
        # Tagalog
        r'parangal', r'pagkilala', r'pinakamahusay na gawi',
        # Thai
        r'รางวัล', r'การยอมรับ', r'แนวปฏิบัติที่ดีที่สุด'
    ],
   'Renewable': [
        # English
        r"\brenewable\b", r"solar", r"wind", r"hydro", r"geothermal", r"biomass",
        # Bahasa Indonesia
        r"energi terbarukan", r"tenaga surya", r"angin", r"air", r"panas bumi", r"biomassa",
        # Bahasa Melayu
        r"tenaga boleh baharu", r"tenaga suria", r"angin", r"hidro", r"geoterma", r"biomassa",
        # Tagalog
        r"renewable", r"solar", r"hangin", r"hidro", r"biomassa",
        # Thai
        r"พลังงานหมุนเวียน", r"พลังงานทดแทน", r"พลังงานแสงอาทิตย์", r"พลังงานลม", r"พลังน้ำ", r"พลังงานความร้อนใต้พิภพ", r"ชีวมวล"
    ],
    'Emission_Carbon': [
        # English
        r"emission", r"carbon", r"greenhouse", r"CO2",
        # Bahasa Indonesia
        r"emisi", r"karbon", r"gas rumah kaca", r"CO2",
        # Bahasa Melayu
        r"pelepasan", r"karbon", r"gas rumah hijau",
        # Tagalog
        r"emisyon", r"carbon", r"greenhouse gas",
        # Thai
        r"การปล่อยก๊าซ", r"คาร์บอน", r"ก๊าซเรือนกระจก", r"การปล่อย CO2"
    ],
    'Efficiency': [
        # English
        r"efficiency", r"energy saving", r"conservation",
        # Bahasa Indonesia
        r"efisiensi", r"hemat energi", r"konservasi energi",
        # Bahasa Melayu
        r"kecekapan tenaga", r"penjimatan tenaga", r"pemuliharaan tenaga",
        # Tagalog
        r"episyente", r"pagtitipid ng enerhiya",
        # Thai
        r"ประสิทธิภาพพลังงาน", r"การใช้พลังงานอย่างมีประสิทธิภาพ", r"การประหยัดพลังงาน"
    ],
    'Waste_Pollution': [
        # English
        r"waste", r"pollution", r"plastic", r"garbage", r"landfill",
        # Bahasa Indonesia
        r"limbah", r"sampah", r"polusi", r"TPA",
        # Bahasa Melayu
        r"sisa", r"sampah", r"pencemaran",
        # Tagalog
        r"basura", r"polusyon", r"plastik",
        # Thai
        r"ของเสีย", r"ขยะ", r"มลพิษ", r"กาก", r"หลุมฝังกลบ"
    ],
    'Policy_Law': [
        # English
        r"policy", r"law", r"regulation", r"act", r"agreement",
        # Bahasa Indonesia
        r"kebijakan", r"undang", r"peraturan", r"UU", r"kesepakatan",
        # Bahasa Melayu
        r"dasar", r"undang-undang", r"peraturan",
        # Tagalog
        r"patakaran", r"batas", r"regulasyon",
        # Thai
        r"นโยบาย", r"กฎหมาย", r"ข้อบังคับ", r"พระราชบัญญัติ", r"ข้อตกลง"
    ],
    'Green_Port_Tech': [
    r'green port', r'eco[- ]?port', r'smart port', r'digital port', r'ecoport',
    r'pelabuhan hijau', r'pelabuhan pintar', r'berdeng pantalan', r'matalinong pantalan',
    r'ท่าเรือสีเขียว', r'ท่าเรืออัจฉริยะ'
],
'Alt_Fuels_Tech': [
    r'LNG', r'liquefied natural gas', r'methanol', r'ammonia', r'hydrogen', r'H2',
    r'biofuel', r'biodiesel', r'e-fuel', r'e-methanol', r'e-ammonia',
    r'CCUS', r'carbon capture', r'fuel cell',
    r'bahan bakar alternatif', r'hidrogen', r'metanol', r'amonia',
    r'alternatibong gasolina', r'hidrogeno', r'метанол',
    r'เชื้อเพลิงทางเลือก', r'ไฮโดรเจน', r'เมทานอล', r'แอมโมเนีย'
]

}

# normalize regex
for k in pillars:
    pillars[k] = [re.compile(pat, flags=re.I) for pat in pillars[k]]


# ----------------------------
# Helper: fetch and extract text from URL (HTML or PDF)
# ----------------------------

def fetch_text(url, timeout=20):
    try:
        r = requests.get(url, timeout=timeout)
        ct = r.headers.get('Content-Type','')
        if 'application/pdf' in ct or url.lower().endswith('.pdf'):
            # parse PDF
            with pdfplumber.open(io.BytesIO(r.content)) as pdf:
                pages = [p.extract_text() or '' for p in pdf.pages]
            text = '\n'.join(pages)
        else:
            soup = BeautifulSoup(r.text, 'html.parser')
            # remove scripts/styles
            for s in soup(['script','style','noscript']):
                s.decompose()
            text = soup.get_text(separator=' ', strip=True)
        return text
    except Exception as e:
        print(f"Failed to fetch {url}: {e}")
        return ''

# ----------------------------
# Crawl sources and build evidence counts per country
# ----------------------------
countries = ['Indonesia','Malaysia','Thailand','Singapore','Philippines']
years = list(range(2018, 2025))  # 2018..2024

# store raw texts (for review) and hits
raw_texts = {}
evidence = {c: {y: {p:0 for p in pillars} for y in years} for c in countries}

print('Fetching source pages (this may take a while)...')
for country, url_list in sources.items():
    for url in tqdm(url_list, desc=country):
        text = fetch_text(url)
        raw_texts[url] = text
        # naive time detection: find years mentioned in text; if none, assume applies to all recent years
        found_years = set(re.findall(r'20(1[8-9]|2[0-4])', text))
        found_years = {int('20'+y) for y in found_years}
        if not found_years:
            found_years = set(years)
        # match pillar keywords
        for p_name, patterns in pillars.items():
            matches = 0
            for pat in patterns:
                if pat.search(text):
                    matches += 1
            # assign evidence counts to found_years
            for y in found_years:
                if country in countries:
                    evidence[country][y][p_name] += matches
                else:
                    # regional/global -> add to all countries as weak evidence
                    for c in countries:
                        evidence[c][y][p_name] += max(0, matches-0)  # leave as is

# ----------------------------
# Map evidence counts -> ordinal score (0/1/2)
# Heuristic: 0 if 0 matches, 1 if 1-2 unique pattern groups, 2 if >=3
# ----------------------------

def map_to_ordinal(count):
    if count <= 0:
        return 0
    elif count <= 2:
        return 1
    else:
        return 2

rows = []
for c in countries:
    for y in years:
        row = {'country':c, 'year':y}
        for p in pillars:
            row[p] = map_to_ordinal(evidence[c][y][p])
        rows.append(row)

df = pd.DataFrame(rows)

# ----------------------------
# Compute GPRSI
# ----------------------------
component_cols = list(pillars.keys())
df['GPRSI_raw_mean'] = df[component_cols].mean(axis=1)
df['GPRSI_0_100'] = df['GPRSI_raw_mean'] * 50

# Min-max across sample per year-grouping optional
scaler = MinMaxScaler(feature_range=(0,100))
df['sum_score'] = df[component_cols].sum(axis=1)
df['GPRSI_minmax'] = scaler.fit_transform(df[['sum_score']])

# Save CSV for review
out_csv = 'gprsi_country_year_2018_2024.csv'
df.to_csv(out_csv, index=False)
print('\nSaved GPRSI CSV to', out_csv)

# ----------------------------
# Visualization: trend lines for each country
# ----------------------------
plt.figure(figsize=(10,5))
for c in countries:
    sub = df[df['country']==c]
    plt.plot(sub['year'], sub['GPRSI_0_100'], marker='o', label=c)
plt.title('GPRSI (0-100) by country — 2018-2024 (heuristic scrape)')
plt.xlabel('Year')
plt.ylabel('GPRSI (0-100)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# ----------------------------
# Component-level normalization for radar/parallel plots
# ----------------------------
latest_year = max(years)
latest_df = df[df['year'] == latest_year]

if latest_df.empty:
    latest_year = max(df['year'])
    latest_df = df[df['year'] == latest_year]

# Build normalized_df for plotting
normalized_rows = []
for _, row in latest_df.iterrows():
    c = row['country']
    for comp in component_cols:
        normalized_rows.append({
            "Country": c,
            "Component": comp,
            "Raw": row[comp],
            "Normalized_Score": row[comp] / 2  # since scores are ordinal 0-2 → normalize to 0–1
        })

normalized_df = pd.DataFrame(normalized_rows)

# ----------------------------
# Plot normalized component comparison
# ----------------------------
plt.figure(figsize=(14,6))

for country in normalized_df['Country'].unique():
    country_data = normalized_df[normalized_df['Country'] == country]
    plt.plot(country_data['Component'], country_data['Normalized_Score'], marker='o', label=country)

plt.title(f'Normalized components comparison (0-1) — {latest_year}')
plt.xlabel('Components')
plt.ylabel('Normalized Score (0-1)')
plt.xticks(rotation=45, ha='right', fontsize=9)
plt.legend()
plt.tight_layout()
plt.show()

# ----------------------------
# Radar (Spider) Chart for normalized component comparison
# ----------------------------
from math import pi

# Pivot the data for radar chart
pivot_df = normalized_df.pivot(index="Country", columns="Component", values="Normalized_Score")

# Ensure consistent component order
components = list(pivot_df.columns)
N = len(components)

# Angle for each component
angles = [n / float(N) * 2 * pi for n in range(N)]
angles += angles[:1]  # complete the loop

# Radar chart
plt.figure(figsize=(8,8))
ax = plt.subplot(111, polar=True)

# Plot each country
for i, country in enumerate(pivot_df.index):
    values = pivot_df.loc[country].tolist()
    values += values[:1]  # repeat first value
    ax.plot(angles, values, label=country, linewidth=2)
    ax.fill(angles, values, alpha=0.1)  # transparent fill

# Add component labels
plt.xticks(angles[:-1], components, fontsize=9)

# Y-labels (radial ticks)
ax.set_rlabel_position(30)
plt.yticks([0.25,0.5,0.75,1.0], ["0.25","0.5","0.75","1"], color="grey", size=8)
plt.ylim(0,1)

plt.title(f"Radar chart of normalized components — {latest_year}", size=14, weight="bold", y=1.1)
plt.legend(loc='upper right', bbox_to_anchor=(1.2, 1.1))
plt.tight_layout()
plt.show()

# Radar chart
plt.figure(figsize=(8,8))
ax = plt.subplot(111, polar=True)

# Plot each country, highlight Indonesia
for i, country in enumerate(pivot_df.index):
    values = pivot_df.loc[country].tolist()
    values += values[:1]  # repeat first value

    if country == "Indonesia":
        ax.plot(angles, values, label=country, linewidth=3, color="gold")
        ax.fill(angles, values, alpha=0.25, color="gold")
    else:
        ax.plot(angles, values, label=country, linewidth=1.5, alpha=0.7)
        ax.fill(angles, values, alpha=0.1)

# Add component labels
plt.xticks(angles[:-1], components, fontsize=9)

# Y-labels (radial ticks)
ax.set_rlabel_position(30)
plt.yticks([0.25,0.5,0.75,1.0], ["0.25","0.5","0.75","1"], color="grey", size=8)
plt.ylim(0,1)

plt.title(f"Radar chart of normalized components — {latest_year}", size=14, weight="bold", y=1.1)

# --- Reorder legend so Indonesia always comes first ---
handles, labels = ax.get_legend_handles_labels()
if "Indonesia" in labels:
    # put Indonesia first, then others
    indonesia_idx = labels.index("Indonesia")
    handles = [handles[indonesia_idx]] + handles[:indonesia_idx] + handles[indonesia_idx+1:]
    labels = ["Indonesia"] + labels[:indonesia_idx] + labels[indonesia_idx+1:]

plt.legend(handles, labels, loc='upper right', bbox_to_anchor=(1.2, 1.1))

plt.tight_layout()
plt.show()


# ----------------------------
# Quick automated insights (text)
# ----------------------------
insights = []
# latest snapshot ranking
snapshot = latest_df.sort_values('GPRSI_0_100', ascending=False)
top = snapshot.iloc[0]
insights.append(f"Top performer in {latest_year} by heuristic GPRSI: {top['country']} ({top['GPRSI_0_100']:.1f}/100)")
insights.append(f"Mean GPRSI in {latest_year}: {snapshot['GPRSI_0_100'].mean():.1f}/100")
insights.append('Component variation: view the parallel coordinates plot for pillar-level differences.')

print('\n=== Automated Insights ===')
for s in insights:
    print('-', s)

# ----------------------------
# Save raw texts for manual review
# ----------------------------
import json
with open('raw_texts.json','w', encoding='utf-8') as f:
    json.dump({k:v for k,v in raw_texts.items()}, f)
print('Saved raw texts to raw_texts.json')

# End of notebook
print('\nNotebook run complete. PLEASE review raw_texts.json and the CSV to manually validate codings.')
```

```python
# ----------------------------
# Follow-up cell: Contextual MAB baseline (LinUCB) using GPRSI CSV
# ----------------------------
# Requirements: scikit-learn, matplotlib, pandas, numpy (already in your Colab notebook)
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import math
import random
from collections import defaultdict

# ----------------------------
# Config / hyperparams
# ----------------------------
CSV_PATH = 'gprsi_country_year_2018_2024.csv'  # produced by your previous cell
RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)
random.seed(RANDOM_SEED)

# Bandit hyperparams
LINUCB_ALPHA = 1.0   # exploration parameter
N_ROUNDS = 2000      # number of bandit timesteps to simulate
VERBOSE = True

# ----------------------------
# Load data
# ----------------------------
try:
    df = pd.read_csv(CSV_PATH)
except FileNotFoundError:
    raise FileNotFoundError(f"Could not find {CSV_PATH}. Run the scraping cell first and ensure the CSV is saved.")

# Basic sanity: require columns
required_cols = {'country','year','GPRSI_0_100'}
if not required_cols.issubset(set(df.columns)):
    raise ValueError(f"CSV must contain columns {required_cols}. Found {df.columns.tolist()}")

# Sort & compute delta to next year (reward proxy)
df = df.sort_values(['country','year']).reset_index(drop=True)
df['GPRSI_next'] = df.groupby('country')['GPRSI_0_100'].shift(-1)
df = df.dropna(subset=['GPRSI_next']).copy()
df['delta'] = df['GPRSI_next'] - df['GPRSI_0_100']  # observed next-year change (can be negative)

# ----------------------------
# Define arms
# arms are policy levers (the pillars) — the bandit chooses which pillar to invest in
# Use the pillar column names from your dataset (subset of df.columns)
# ----------------------------
# detect pillar columns by presence in df
possible_pillars = ['standards','economic_instruments','infrastructure','MRV_transparency','recognition']
arms = [p for p in possible_pillars if p in df.columns]
if len(arms) == 0:
    raise ValueError("No pillar columns found in CSV. Ensure pillar columns (e.g., 'standards') exist.")

K = len(arms)
if VERBOSE:
    print(f"Detected arms (pillars): {arms}")

# ----------------------------
# Build context vectors
# Features: one-hot country, normalized year, current pillar scores, sum_score (if present)
# ----------------------------
# Country one-hot
countries = df['country'].unique().tolist()
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
country_ohe = ohe.fit_transform(df[['country']])

# Year normalized
year_scaler = StandardScaler()
year_norm = year_scaler.fit_transform(df[['year']])

# Pillar scores (as-is)
pillar_feats = df[arms].fillna(0).values.astype(float)

# optional sum_score if present
if 'sum_score' in df.columns:
    sum_score = df[['sum_score']].fillna(0).values.astype(float)
    X_base = np.hstack([country_ohe, year_norm, pillar_feats, sum_score])
else:
    X_base = np.hstack([country_ohe, year_norm, pillar_feats])

# Standardize contexts
ctx_scaler = StandardScaler()
X = ctx_scaler.fit_transform(X_base)

# Keep contexts and rewards
contexts = X  # each row is a context
rewards_observed = df['delta'].values  # used to train surrogate
n_examples = contexts.shape[0]
if VERBOSE:
    print(f"Prepared {n_examples} contexts (country-year observations). Context dimension = {contexts.shape[1]}")

# ----------------------------
# Train surrogate reward model
# We train a regressor that predicts delta (next-year GPRSI change) given (context + action)
# This model will act as the environment simulator: reward = surrogate.predict(context,action) + small_noise
# ----------------------------
# Create action one-hot features appended to context for training
action_ohe = np.eye(K)  # identity for K arms
# expand training data by making K copies for each example, one per possible action.
X_aug = np.repeat(contexts, K, axis=0)
actions_aug = np.tile(np.arange(K), n_examples)
action_features = action_ohe[actions_aug]
X_aug_full = np.hstack([X_aug, action_features])
y_aug = np.repeat(rewards_observed, K)  # naive: we use same observed delta for all hypothetical actions
# NOTE: because we don't have observed rewards per action, this surrogate is limited.
# To make the surrogate responsive to actions, we add a small synthetic signal:
# We'll inject a weak signal that actions aligned with pillar presence produce slightly higher reward.
# Construct a synthetic action effect based on current pillar score for that arm:
action_effect = np.zeros_like(y_aug)
for i in range(n_examples):
    base_idx = i*K
    for a in range(K):
        # use pillar score in original data for this arm as a proxy: higher existing pillar -> higher marginality? it's heuristic
        pillar_score = pillar_feats[i, a] if a < pillar_feats.shape[1] else 0.0
        action_effect[base_idx + a] = 0.2 * pillar_score  # small boost proportional to pillar_presence

y_aug = y_aug + action_effect

# Train regressor
rf = RandomForestRegressor(n_estimators=200, random_state=RANDOM_SEED, max_depth=6)
rf.fit(X_aug_full, y_aug)
if VERBOSE:
    print("Trained surrogate RandomForestRegressor (env simulator).")

# environment function: given context vector (1d) and chosen arm index -> predicted reward (float)
def env_reward_predict(context_vec, arm_idx, noise_std=0.05):
    a_onehot = np.zeros(K)
    a_onehot[arm_idx] = 1.0
    x_full = np.concatenate([context_vec, a_onehot]).reshape(1, -1)
    pred = rf.predict(x_full)[0]
    # small gaussian noise to make simulation stochastic
    return pred + np.random.normal(scale=noise_std)

# ----------------------------
# LinUCB implementation (per-arm linear models)
# Classic LinUCB (A_a = Dxd, b_a = Dx1) where context is d-dim vector
# ----------------------------
d = contexts.shape[1]
# initialize A and b for each arm
A = [np.eye(d) for _ in range(K)]
b = [np.zeros(d) for _ in range(K)]
theta = [np.zeros(d) for _ in range(K)]

def lincb_select_arm(context_vec, alpha=LINUCB_ALPHA):
    p_vals = np.zeros(K)
    for a in range(K):
        A_inv = np.linalg.inv(A[a])
        theta[a] = A_inv.dot(b[a])
        mean = theta[a].dot(context_vec)
        sigma = np.sqrt(context_vec.dot(A_inv).dot(context_vec))
        p_vals[a] = mean + alpha * sigma
    # choose arm with highest UCB
    return int(np.argmax(p_vals)), p_vals

# ----------------------------
# Simulation loop
# We'll sample contexts randomly from our observed contexts and simulate decisions for N_ROUNDS steps.
# Compare LinUCB to Random baseline and to an 'oracle' that always picks best arm per surrogate.
# ----------------------------
cumulative_reward_lincb = []
cumulative_reward_random = []
cumulative_reward_oracle = []

total_reward_lincb = 0.0
total_reward_random = 0.0
total_reward_oracle = 0.0

# track chosen counts
counts_lincb = np.zeros(K, dtype=int)
counts_random = np.zeros(K, dtype=int)

# precompute oracle choices (which arm gives highest predicted reward per context)
oracle_choice = []
for i in range(n_examples):
    ctx = contexts[i]
    preds = [env_reward_predict(ctx, a, noise_std=0.0) for a in range(K)]
    oracle_choice.append(int(np.argmax(preds)))

# run rounds
for t in range(N_ROUNDS):
    # sample a random context index (simulate encountering a country-year)
    idx = np.random.randint(0, n_examples)
    ctx = contexts[idx]

    # LinUCB selection
    arm_l, pvals = lincb_select_arm(ctx, alpha=LINUCB_ALPHA)
    r_l = env_reward_predict(ctx, arm_l)
    # update LinUCB parameters for chosen arm
    A[arm_l] += np.outer(ctx, ctx)
    b[arm_l] += r_l * ctx
    total_reward_lincb += r_l
    counts_lincb[arm_l] += 1
    cumulative_reward_lincb.append(total_reward_lincb)

    # Random baseline
    arm_r = np.random.randint(0, K)
    r_r = env_reward_predict(ctx, arm_r)
    total_reward_random += r_r
    counts_random[arm_r] += 1
    cumulative_reward_random.append(total_reward_random)

    # Oracle (cheating; uses surrogate to pick best)
    arm_o = oracle_choice[idx]
    r_o = env_reward_predict(ctx, arm_o)
    total_reward_oracle += r_o
    cumulative_reward_oracle.append(total_reward_oracle)

# ----------------------------
# Plot cumulative rewards
# ----------------------------
plt.figure(figsize=(10,6))
plt.plot(cumulative_reward_lincb, label=f"LinUCB (alpha={LINUCB_ALPHA})")
plt.plot(cumulative_reward_random, label="Random")
plt.plot(cumulative_reward_oracle, label="Oracle (surrogate best)")
plt.xlabel("Round")
plt.ylabel("Cumulative reward (simulated Δ GPRSI_0_100)")
plt.title("Contextual Bandit Baselines — cumulative reward")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# ----------------------------
# Summary statistics
# ----------------------------
print("\n--- Summary ---")
print(f"Rounds simulated: {N_ROUNDS}")
print(f"LinUCB total reward: {total_reward_lincb:.3f}")
print(f"Random total reward: {total_reward_random:.3f}")
print(f"Oracle total reward: {total_reward_oracle:.3f}")
print("\nLinUCB chosen counts per arm:")
for i,a in enumerate(arms):
    print(f"  {a}: {counts_lincb[i]}")
print("\nRandom chosen counts per arm:")
for i,a in enumerate(arms):
    print(f"  {a}: {counts_random[i]}")

# Optionally: save results to CSV
out_hist = pd.DataFrame({
    'round': np.arange(1, N_ROUNDS+1),
    'cum_reward_lincb': cumulative_reward_lincb,
    'cum_reward_random': cumulative_reward_random,
    'cum_reward_oracle': cumulative_reward_oracle
})
out_hist.to_csv('mab_simulation_history.csv', index=False)
print("\nSaved simulation history to mab_simulation_history.csv")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# 1. Define the Action Space
# Based on your pillars:
arms = ['technological_infrastructure', 'advanced_autonomous_technologies',
        'interoperability_communication', 'onshore_operation_centers',
        'cybersecurity_resilience', 'regulation_procedures']
K = len(arms)

# 2. Prepare Context (x_t)
def prepare_context(df):
    # One-hot Country
    ohe = OneHotEncoder(sparse_output=False)
    country_feats = ohe.fit_transform(df[['country']])

    # Scale Year and Pillar Scores
    scaler = StandardScaler()
    other_feats = scaler.fit_transform(df[['year'] + arms])

    X = np.hstack([country_feats, other_feats])
    return X, ohe, scaler

# 3. LinUCB Agent (The "RL" Brain)
class LinUCBAgent:
    def __init__(self, context_dim, n_arms, alpha=1.0):
        self.d = context_dim
        self.K = n_arms
        self.alpha = alpha
        # Learning Update Variables
        self.A = [np.eye(self.d) for _ in range(self.K)]  # Covariance
        self.b = [np.zeros((self.d, 1)) for _ in range(self.K)] # Reward vector

    def select_arm(self, x_t):
        x_t = x_t.reshape(-1, 1)
        p = np.zeros(self.K)
        for a in range(self.K):
            A_inv = np.linalg.inv(self.A[a])
            theta_hat = A_inv @ self.b[a]
            # Learning Update Rule (UCB Calculation)
            variance = np.sqrt(x_t.T @ A_inv @ x_t)
            p[a] = theta_hat.T @ x_t + self.alpha * variance
        return np.argmax(p)

    def update(self, arm, x_t, reward):
        x_t = x_t.reshape(-1, 1)
        # Learning Update Rule (Updating A and b)
        self.A[arm] += x_t @ x_t.T
        self.b[arm] += reward * x_t

# 4. Simulation Environment (The Reward Function)
# Rewards represent the "gain" in index score.
def get_reward(context_vec, arm_idx, rf_model):
    # This uses your trained RandomForest surrogate to simulate how the real world
    # would respond to a policy action in a specific country/context.
    action_vec = np.zeros(K)
    action_vec[arm_idx] = 1
    feat_full = np.concatenate([context_vec, action_vec]).reshape(1, -1)
    return rf_model.predict(feat_full)[0] + np.random.normal(0, 0.05)

print("RL Framework initialized: Action space, Context, and Update rules ready.")
```

```python
# ============================================================
# SMART PORT READINESS INDEX (SPRI) - REINFORCEMENT LEARNING SIMULATION
# Framework: Contextual Multi-Armed Bandit (LinUCB)
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import OneHotEncoder, StandardScaler
import random

# 1. CONFIGURATION & HYPERPARAMETERS
CSV_PATH = 'spri_2018_2024.csv'  # Ensure this matches your filename in Colab
RANDOM_SEED = 42
LINUCB_ALPHA = 1.0   # Exploration parameter
N_ROUNDS = 2000      # Total decision rounds
np.random.seed(RANDOM_SEED)
random.seed(RANDOM_SEED)

# 2. LOAD & PREPROCESS DATA
try:
    df = pd.read_csv(CSV_PATH)
    print(f"Successfully loaded {CSV_PATH}")
except FileNotFoundError:
    print(f"ERROR: {CSV_PATH} not found. Please upload the file to Colab.")
    raise

# Define the Action Space (The 6 SPRI Pillars)
arms = [
    'technological_infrastructure',
    'advanced_autonomous_technologies',
    'interoperability_communication',
    'onshore_operation_centers',
    'cybersecurity_resilience',
    'regulation_procedures'
]
K = len(arms)

# Calculate Reward Proxy (Next-year change in SPRI)
df = df.sort_values(['country', 'year']).reset_index(drop=True)
df['SPRI_next'] = df.groupby('country')['SPRI_0_100'].shift(-1)
df = df.dropna(subset=['SPRI_next']).copy()
df['delta'] = df['SPRI_next'] - df['SPRI_0_100']

# 3. CONTEXT FEATURE ENGINEERING (x_t)
# One-hot encoding for countries
ohe = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
country_ohe = ohe.fit_transform(df[['country']])

# Normalizing year and pillar scores
ctx_scaler = StandardScaler()
X_base = np.hstack([country_ohe, df[['year']].values, df[arms].values])
X_contexts = ctx_scaler.fit_transform(X_base)
context_dim = X_contexts.shape[1]

# 4. ENVIRONMENT SIMULATOR (Surrogate Reward Model)
# We train a model to "act" as the world, telling the agent the reward for its action
X_aug = np.repeat(X_contexts, K, axis=0)
actions_aug = np.tile(np.arange(K), len(X_contexts))
action_features = np.eye(K)[actions_aug]
X_train = np.hstack([X_aug, action_features])
y_train = np.repeat(df['delta'].values, K)

# Add synthetic signal: Investment in a pillar gives a 0.2 boost if pillar is already present
for i in range(len(X_contexts)):
    for a in range(K):
        y_train[i*K + a] += 0.2 * df[arms].iloc[i, a]

rf_env = RandomForestRegressor(n_estimators=100, max_depth=6, random_state=RANDOM_SEED)
rf_env.fit(X_train, y_train)

def get_env_reward(context_vec, arm_idx):
    a_onehot = np.zeros(K)
    a_onehot[arm_idx] = 1.0
    feat = np.concatenate([context_vec, a_onehot]).reshape(1, -1)
    return rf_env.predict(feat)[0] + np.random.normal(0, 0.05)

# 5. THE RL AGENT (LinUCB Algorithm)
class LinUCBAgent:
    def __init__(self, d, K, alpha):
        self.A = [np.eye(d) for _ in range(K)]
        self.b = [np.zeros((d, 1)) for _ in range(K)]
        self.alpha = alpha
        self.K = K

    def select_arm(self, x_t):
        x_t = x_t.reshape(-1, 1)
        p = np.zeros(self.K)
        for a in range(self.K):
            A_inv = np.linalg.inv(self.A[a])
            theta = A_inv @ self.b[a]
            uncertainty = self.alpha * np.sqrt(x_t.T @ A_inv @ x_t)
            p[a] = theta.T @ x_t + uncertainty
        return np.argmax(p)

    def update(self, arm, x_t, reward):
        x_t = x_t.reshape(-1, 1)
        self.A[arm] += x_t @ x_t.T
        self.b[arm] += reward * x_t

# 6. EXECUTE SIMULATION
agent = LinUCBAgent(context_dim, K, LINUCB_ALPHA)
hist_lincb, hist_random, hist_oracle = [], [], []
cum_lincb, cum_random, cum_oracle = 0, 0, 0
counts = np.zeros(K)

for t in range(N_ROUNDS):
    idx = np.random.randint(0, len(X_contexts))
    x_t = X_contexts[idx]

    # LinUCB Policy
    arm_l = agent.select_arm(x_t)
    reward_l = get_env_reward(x_t, arm_l)
    agent.update(arm_l, x_t, reward_l)
    cum_lincb += reward_l
    counts[arm_l] += 1

    # Random Policy (Baseline)
    reward_r = get_env_reward(x_t, np.random.randint(0, K))
    cum_random += reward_r

    # Oracle Policy (Perfect Knowledge)
    all_rewards = [get_env_reward(x_t, a) for a in range(K)]
    cum_oracle += max(all_rewards)

    hist_lincb.append(cum_lincb)
    hist_random.append(cum_random)
    hist_oracle.append(cum_oracle)

# 7. RESULTS & VISUALIZATION
plt.figure(figsize=(10, 5))
plt.plot(hist_lincb, label=f"LinUCB (Alpha={LINUCB_ALPHA})", linewidth=2)
plt.plot(hist_random, label="Random Policy", linestyle='--')
plt.plot(hist_oracle, label="Oracle Policy", linestyle=':')
plt.title("SPRI Predictive Utility: Cumulative Reward Comparison")
plt.xlabel("Simulation Round")
plt.ylabel("Cumulative Reward (Δ SPRI)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("\n--- Simulation Summary ---")
for i, arm in enumerate(arms):
    print(f"Pillar: {arm:35} | Times Chosen: {int(counts[i])}")
```
